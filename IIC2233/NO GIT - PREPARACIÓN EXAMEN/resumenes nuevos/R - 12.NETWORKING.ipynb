### <span style="color:#ff66cc; font-weight:bold;">1) SERIALIZACIÓN DE OBJETOS</span> 

Consiste en transformar un objeto en una serie de `bytes`. Esto permite que un objeto pueda trabajarse en diferentes programas, entre diferentes lenguajes.

### <span style="color:#ff66cc; font-weight:bold;">2) JSON</span> 

Formato de texto interpretado por muchos lenguajes. Ese sirve como serialización para traspaso de información.

Se debe importar `json`. Código:

---

<span style="color:#00cc66; font-weight:bold;">1. `dumps`:</span> Serializa un objeto y lo guarda como `str` en formato `JSON`.

<span style="color:#00cc66; font-weight:bold;">2. `loads`: </span> Deserealiza un `str` en formato `JSON` y lo carga como un objeto de `Python`.

<span style="color:#00cc66; font-weight:bold;">3. `dump`: </span> Serializa un objeto en un archivo.

<span style="color:#00cc66; font-weight:bold;">2. `loads`: </span> Deserealiza un archivo en formato `JSON` y lo carga como un objeto de `Python`.

---

### <span style="color:#ff66cc; font-weight:bold;">3) SERIALIZACIÓN DE OBJETOS PERSONALIZADOS</span> 

No todo contenido puede serializarse inmediatamente. Por ejemplo los `objetos` no pueden serializarse, por lo que se debe modificar su contenido:

````python
import json

lista = ["a", 1, 3, "hola"]
lista_serializada = json.dumps(lista)

print(f"Resultado serialización: {lista_serializada}")
print(f"Tipo de versión serializada: {type(lista_serializada)}")

lista_deserializada = json.loads(lista_serializada)
print(f"Resultado deserialización: {lista_deserializada}")

print()
print(f"Lista original: {lista}")
print(f"Lista deserializada: {lista_deserializada}")
print(f"¿Las listas son iguales? {lista == lista_deserializada}")
print(f"¿Las listas son el mismo objeto? {lista is lista_deserializada}")
````

Los que no tienen problema son `int`, `str`, `float`, `dict`, `bool`, `list`, `tuple` y `None`. 

### <span style="color:#ff66cc; font-weight:bold;">4) DESERIALIZACIÓN EN `JSON`</span> 

Tambien se puede `deserializar` de forma personalizada usando `object_hook`:

````python
def funcion_lista_hook(diccionario: dict):
    return [(key, value) for key, value in diccionario.items()]


json_string = '{"nombre": "Luna", "edad": 8, "especie": "Gato", "ano_nacimiento": 2017}'
datos = json.loads(json_string, object_hook=funcion_lista_hook)

print(datos)
````

## <span style="color:#ff5555; font-weight:bold;">2) NETWORKING: CONCEPTOS BÁSICOS</span> 

### <span style="color:#ff66cc; font-weight:bold;">1) COMUNICACIÓN ENTRE COMPUTADORES</span> 

Para que 2 máquinas se comuniquen necesitan:

<span style="color:#00cc66; font-weight:bold;">1. Un medio</span>  
<span style="color:#00cc66; font-weight:bold;">2. Destinatario. Como encontrarlo.</span>  
<span style="color:#00cc66; font-weight:bold;">3. Un protocolo de transmisión de mensajes</span>  
<span style="color:#00cc66; font-weight:bold;">4. Un protocolo de aplicación. Orden y formato de mensajes.</span> 

Para identificar cada máquina <span style="color:#f1c40f; font-weight:bold;">(host)</span> se usan etiquetas llamadas `dirección ip`. Es practicamente única en toda la red y permite que nos comuniquemos a través de internet.  

### <span style="color:#ff66cc; font-weight:bold;">2) IP</span> 

Existen 2 tipos de `ip`:

<span style="color:#f1c40f; font-weight:bold;">1. IPv4:</span> Cada `ip` con 32 bits (4 bytes) y en total 2^32 posibilidades de `ip`.

<span style="color:#f1c40f; font-weight:bold;">1. IPv6:</span> Cada `ip` con 128 bits (6 bytes) y en total 2^128 posibilidades de `ip`.

### <span style="color:#ff66cc; font-weight:bold;">3) PORTS</span> 

Un `host` posee diversos `ports`. Puertas que permiten la ejecución de varios programas. Eso se usa para identificar que proceso se esta ejecutando, siendo que existen desde el 0 hasta 2^16 - 1 puertos disponibles (en total son `65.536`)

### <span style="color:#ff66cc; font-weight:bold;">4) PROTOCOLOS DE COMUNICACIÓN</span> 

Un mensaje se divide en varios `paquetes` (minimensajes) para que todo llegue bien. Por ello nacen <span style="color:#ff8800; font-weight:bold;">2 protocolos de transporte:</span>

##### <span style="color:#ff8800; font-weight:bold;">1. TCP</span> 

Forma confiable de transmisión de información pero lenta. El `emisor` y `receptor` deben hacer un `handshake` para que ambos esten de acuerdo en cómo hacer las cosas y organizarse. Luego de eso, se continua el protocolo. Para que funcione correctamente depende de 3 cosas para su correcto funcionamiento:


<span style="color:#00cc66; font-weight:bold;">1. Corroborar que cada `paquete` llegue de forma correcta.</span>

<span style="color:#00cc66; font-weight:bold;">2. Solicitar la confirmación al receptor que cada paquete haya llegado bien.</span>

<span style="color:#00cc66; font-weight:bold;">3. Reenviarlo que haya llegado mal.</span> 

##### <span style="color:#ff8800; font-weight:bold;">2. UDP</span> 

Este envia y permite recibir de forma rápida pero es insegura. <span style="color:#ff8800; font-weight:bold;">No se preocupa en realizar un</span> `handshake`. Es más de recibe y chao, ya que no se preocupa que la información se reciba bien.

Cabe decir que la comunicación entre 2 hosts se divide en capas.

### <span style="color:#ff66cc; font-weight:bold;">5) ARQUITECTURA CLIENTE-SERVIDOR</span> 

Una ***arquitectura*** en computación se refiere a como construir un *hardware* o *software* para que cumplan ciertos requisitos. En este caso, `cliente-servidor` busca que 2 o más programas que conecten y que sirvan como un `servidor` y como un `cliente`.

El cliente inicia la conexión entre ellos y el servidor sigue un protocolo. El `cliente` hace solicitudes y el `servidor` decide si las acepta o no según ciertas reglas.

## <span style="color:#ff5555; font-weight:bold;">3) NETWORKING: WEBSERVICES: CONCEPTOS BÁSICOS</span> 

### <span style="color:#ff66cc; font-weight:bold;">1) WEB SERVICES</span> 

Aplicaciones que usan la <span style="color:#ff8800; font-weight:bold;">arquitectura cliente-servidor</span>. Un ejemplo de esto es un <span style="color:#f1c40f; font-weight:bold;">navegador</span> y un <span style="color:#f1c40f; font-weight:bold;">sitio web</span>.  

### <span style="color:#ff66cc; font-weight:bold;">2) API</span> 

Conjunto de normas que tienen funciones y más. Estas sirven para que se ejecute la comunicación ***cliente-servidor***. Si hay un servidor que da info del clima, la API define que datos se pueden pedir, cómo pedirlos y más.

### <span style="color:#ff66cc; font-weight:bold;">3) HTTP</span> 

Protocolo de comunicación ***cliente-servidor***. La mayoria de web services usan `HTTP`. Este usa el protocolo ***request-response***. Este no guarda info de las conexiones.

Usa <span style="color:#f1c40f; font-weight:bold;">métodos y verbos</span> que indican la acción a realizar:

<span style="color:#00cc66; font-weight:bold;">1. GET:</span> Recupera una representación de un recurso.

<span style="color:#00cc66; font-weight:bold;">2. POST:</span> Crea un recurso.

<span style="color:#00cc66; font-weight:bold;">3. PATCH:</span> Modificaciones parciales a un recurso.

<span style="color:#00cc66; font-weight:bold;">4. PUT:</span> Reemplaza un recurso existente.

<span style="color:#00cc66; font-weight:bold;">5. DELETE:</span> Elimina un recurso.

---

También http otorga códigos de estado que nacen de las interacciones cliente-servidor:

<span style="color:#00cc66; font-weight:bold;">200:</span> Todo OK :D

<span style="color:#00cc66; font-weight:bold;">403:</span> Prohibido. La solicitud fue aceptada pero el servidor la rechazó.

<span style="color:#00cc66; font-weight:bold;">404:</span> No encontrado.

<span style="color:#00cc66; font-weight:bold;">500:</span> Error interno del servidor 

Las `request` y `response` tienen un `header` y un `body`.

### <span style="color:#ff66cc; font-weight:bold;">3) PARÁMETROS DE UNA CONSULTA</span>

### <span style="color:#ff66cc; font-weight:bold;">5) AUTENTIFICACIÓN CON HEADERS</span> 

Hay otros métodos siendo `.put()`, `.patch()` y `.delete()`. Suelen modificar una base de datos donde necesitan una autorización previa.

Por ello se usa el parámetro `headers` que permite incluir info en la cabezera de la solicitud.

````python
import requests

# URL de ejemplo (API falsa para pruebas)
url = "https://jsonplaceholder.typicode.com/posts/1"

# Headers con autorización (token de ejemplo)
headers = {
    "Authorization": "Bearer mi-token-secreto-123",
    "Content-Type": "application/json"
}

# ===== GET: Obtener datos (solo lectura, no necesita autorización) =====
print("1. GET - Obtener información:")
response = requests.get(url)
print(f"   Datos obtenidos: {response.json()}\n")


# ===== POST: Crear un nuevo recurso =====
print("2. POST - Crear nuevo post:")
nuevo_post = {
    "title": "Mi nuevo post",
    "body": "Contenido del post",
    "userId": 1
}
response = requests.post(
    "https://jsonplaceholder.typicode.com/posts",
    json=nuevo_post,
    headers=headers  # Incluimos autorización
)
print(f"   Post creado: {response.json()}\n")


# ===== PUT: Reemplazar completamente un recurso =====
print("3. PUT - Reemplazar todo el post:")
post_actualizado = {
    "id": 1,
    "title": "Título completamente nuevo",
    "body": "Cuerpo completamente nuevo",
    "userId": 1
}
response = requests.put(url, json=post_actualizado, headers=headers)
print(f"   Post reemplazado: {response.json()}\n")


# ===== PATCH: Modificar parcialmente un recurso =====
print("4. PATCH - Modificar solo el título:")
cambio_parcial = {
    "title": "Solo cambié el título"
}
response = requests.patch(url, json=cambio_parcial, headers=headers)
print(f"   Post modificado: {response.json()}\n")


# ===== DELETE: Eliminar un recurso =====
print("5. DELETE - Eliminar el post:")
response = requests.delete(url, headers=headers)
print(f"   Estado: {response.status_code} (200 = éxito)")
print(f"   Respuesta: {response.text}\n")


# RESUMEN DE MÉTODOS:
print("=" * 50)
print("RESUMEN:")
print("GET    → Obtener/leer datos")
print("POST   → Crear nuevo recurso")
print("PUT    → Reemplazar completamente")
print("PATCH  → Modificar parcialmente")
print("DELETE → Eliminar")
print("\n¡Todos menos GET suelen necesitar autorización!")
````

### <span style="color:#ff66cc; font-weight:bold;">2) FLASK</span> 

La librería `flask` permite levantar una `API` y crear funciones con métodos (`GET`, `POST`, `PUT`, `DELETE`, `PATCH`) y los endpoints deseados.

##### <span style="color:#ff8800; font-weight:bold;">EJEMPLO</span>    

```python
        from flask import Flask, request
        import random

        app = Flask(__name__)


        @app.route("/", methods=["GET"])
        def hello_world():
            return {"texto": "Hello, World!"}


        @app.route("/numero_aleatorio", methods=["GET", "POST"])
        def numero_aleatorio():
            if request.method == "POST":
                numero = random.randint(0, 6)
                return {"texto": f"Tu número es: {numero}", "método": "POST"}

            numero = random.randint(-4444, -11)
            return {"texto": f"Tu número es: {numero}", "método": "GET"}


        if __name__ == "__main__":
            app.run(host="localhost", port=4444)
```

<span style="color:#00cc66; font-weight:bold;">1. Importamos la librería</span>  

<span style="color:#00cc66; font-weight:bold;">2. Creaumos una instancia de aplicacion con `Flask(__name__).`</span>  

<span style="color:#00cc66; font-weight:bold;">3. Para crear los distintos ***endpoints*** de la api usaremos <span style="color:#ff8800; font-weight:bold;">1)</span>  `@app.route` junto con los métodos, <span style="color:#ff8800; font-weight:bold;">2) </span>  definimos la función, <span style="color:#ff8800; font-weight:bold;">3)</span>  Podemos usar `requests.method` para identificar el método de la función</span>

### <span style="color:#ff66cc; font-weight:bold;">4) ENDPOINTS DINÁMICOS</span> 

Son rutas en una API que poseen varaibles que cambian según la solicitud. Parte de la URL se convierte en un parámetro:

``` python
    @app.route("/hi/<string:username>", methods=["GET"])
    def saludar(username):
        return {"texto": f"Hola {username}!"}


    @app.route("/ruta/<tipo:nombre_variable>", methods=["GET"])
    def funcion(nombre_variable):
        # nombre_variable contiene el valor de la URL
        return {"resultado": nombre_variable}
```

En este caso el username permite que existan varios endpoints
