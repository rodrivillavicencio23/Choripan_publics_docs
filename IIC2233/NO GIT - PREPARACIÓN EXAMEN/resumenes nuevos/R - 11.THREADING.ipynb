#### <span style="color:#ff66cc; font-weight:bold;">4) Threading</span>  

Un ***thread*** es una ejecución de un código dentro de un proceso. Todo código tiene al menos un ***thread*** que sería el ***thread principal***.

Cada ***thread*** funciona de manera independiente, aunque pueden <span style="color:#ff8800; font-weight:bold;">crear otros threads o interactuar entre ellos</span>. 

En `python` estos se turnan. Un poco uno, un poco el otro, y asi pareciera que fuera simultaneo.

Esto sirve para separar funciones de todo un programa por partes independientes, como usar una `GUI` y la lógica de un programa por detrás.

#### <span style="color:#ff66cc; font-weight:bold;">5) Módulo `threading`</span>  

Este se importa al inicio del programa con `import threading`. Este da clases y funciones para trabajar con el:

---

<span style="color:#00cc66; font-weight:bold;">1. Thread(target = función):</span> Crea un hilo nuevo que ejecutará la `función` indicada.

<span style="color:#00cc66; font-weight:bold;">2. start(): </span> Inicia la ejecución del hilo.

<span style="color:#00cc66; font-weight:bold;">3. join():</span> Bloquea el hilo que se especifique hasta que otro hilo termine.

<span style="color:#00cc66; font-weight:bold;">4. is:alive():</span> Indica si un hilo esta vivo.

``` python
        import threading

        def contar_diez_ovejas() -> None:
            print("Tengo sueño...")
            for numero in range(1, 11):
                print(f"({numero} oveja{'s' if numero > 1 else ''})")
            print("A dormir...")

        mi_hilo = threading.Thread(target=contar_diez_ovejas)

        saludar()
```

#### <span style="color:#ff66cc; font-weight:bold;">7) Threads con clases</span>  

Al igual que con funciones, se pueden usar clases para los `threads` incluso hacerles <span style="color:#f1c40f; font-weight:bold;">override</span> a la clase de `Thread` del módulo `threading`:

````python
import threading
import time

class MiHilo(threading.Thread):
    def __init__(self, tiempo, name):
        super().__init__(name=name)
        self.tiempo_espera = tiempo
    
    def run(self):
        for i in range(3):
            print(f"Hilo {self.name} ejecutando {i}")
            time.sleep(self.tiempo_espera)
        print(f"Hilo {self.name} ha terminado.")


h1 = MiHilo(tiempo=3, name="HILO_1")
h2 = MiHilo(tiempo=1, name="HILO_2")
h3 = MiHilo(tiempo=2, name="HILO_3")

h1.start()
h2.start()

print("-" * 20, f"hilo 1 esta : {h1.is_alive()}")

h1.join() 

print("-" * 20, f"hilo 1 esta : {h1.is_alive()}")

h3.start()

h2.join()
h3.join()

print("\n--- Se terminó todo exitosamente ---")
````

#### <span style="color:#ff66cc; font-weight:bold;">8) Timers</span>  

Permite que se ejecute un `thread` luego de cierto tiempo sin detener el flujo:

````python
import threading

def mostrar_mensaje(texto):
    print(texto)

timer = threading.Timer(3.0, mostrar_mensaje, args=("Hola luego de 3 segundos",))
print("Empezó y esta esperando 3 segundos")
timer.start()
timer.join()
print("Termino")
````

#### <span style="color:#ff66cc; font-weight:bold;">2) SINCRONIZACIÓN</span>  

##### <span style="color:#00cc66; font-weight:bold;">LOCK</span> 

Se utiliza para que solo un `thread` pueda acceder a una sección <span style="color:#ff8800; font-weight:bold;">crítica</span> a la vez. Osea que el `lock` permite que se usen variables compartidas pero de buena forma.

Este puede estar bloqueado o desbloqueado (Inicialmente esta desbloqueado). Para usar `lock`, un <span style="color:#f1c40f; font-weight:bold;">thread</span> debe hacer `acquire()`, mientras los otros threads tendran que esperar hasta que el que lo bloqueó haga <span style="color:#f1c40f; font-weight:bold;">release()</span>.

````python
import threading


class Contador: 
    def __init__(self) -> None:
        self.valor = 0
        self.lock = threading.Lock()

        
def sumador(contador: Contador) -> None:
    for _ in range(10**6):
        contador.lock.acquire()  # Bloquea el acceso
        try:
            temp = contador.valor
            temp += 1
            contador.valor = temp
        finally:
            contador.lock.release()  # Libera el acceso


contador = Contador()

threads = []
for _ in range(5):
    t = threading.Thread(target=sumador, args=(contador,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Resultado: {contador.valor}")
print(f"Debería ser: {5 * 10**6}")
print(contador.valor == 5 * 10**6)
````

Cabe aclarar que el lock debe ser compartido, sino no servirá de nada xd.

##### <span style="color:#00cc66; font-weight:bold;">LOCK Y CONTEXT MANAGERS</span> 

Tambien se puede hacer locks con `with`, sin necesidad de liberarlo (no de forma explicita).

````python
import threading


class Contador: 
    def __init__(self) -> None:
        self.valor = 0
        self.lock = threading.Lock()

        
def sumador(contador: Contador) -> None:
    for _ in range(10**6):
        with contador.lock:  # Automáticamente hace acquire() y release()
            temp = contador.valor
            temp += 1
            contador.valor = temp


contador = Contador()

threads = []
for _ in range(5):
    t = threading.Thread(target=sumador, args=(contador,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print(f"Resultado: {contador.valor}")
print(f"Debería ser: {5 * 10**6}")
print(contador.valor == 5 * 10**6)
````

#### <span style="color:#ff66cc; font-weight:bold;">3) SEÑALES</span>  

Aunque tengamos `join()` hay otra forma para que un `thread` reaccione a otro, esto usando <span style="color:#ff8800; font-weight:bold;">Event</span>. Posee caracter binario (bool) para que un hilo pueda avanzar.

Un <span style="color:#ff8800; font-weight:bold;">event()</span> tiene:

<span style="color:#00cc66; font-weight:bold;">1. .set():</span> activa el evento.

<span style="color:#00cc66; font-weight:bold;">2. .wait():</span> hace que el hilo espere hasta que se active el evento.

<span style="color:#00cc66; font-weight:bold;">3. .clear():</span> coloca el evento en false.

<span style="color:#00cc66; font-weight:bold;">4. .is_set():</span> consulta si el evento esta activado.

````python
import threading
import time


def trabajador(nombre: str, evento: threading.Event):
    print(f"{nombre}: Esperando la señal...")
    evento.wait()  # Espera hasta que el evento se active
    print(f"{nombre}: ¡Señal recibida! Empezando a trabajar")
    time.sleep(1)
    print(f"{nombre}: Trabajo terminado")


def controlador(evento: threading.Event):
    print("Controlador: Preparando todo...")
    time.sleep(3)
    print("Controlador: ¡Activando señal!")
    evento.set()  # Activa el evento (libera todos los wait())
    
    time.sleep(2)
    print(f"Controlador: ¿Evento activo? {evento.is_set()}")
    
    print("Controlador: Limpiando señal")
    evento.clear()  # Desactiva el evento
    print(f"Controlador: ¿Evento activo? {evento.is_set()}")


# Crear el evento
evento = threading.Event()

print(evento.is_set()) 

# Crear threads
t1 = threading.Thread(target=trabajador, args=("Trabajador-1", evento))
t2 = threading.Thread(target=trabajador, args=("Trabajador-2", evento))
t3 = threading.Thread(target=trabajador, args=("Trabajador-3", evento))
tc = threading.Thread(target=controlador, args=(evento,))

# Iniciar threads
t1.start()
t2.start()
t3.start()
tc.start()

# Esperar a que terminen
t1.join()
t2.join()
t3.join()
tc.join()

print("Programa terminado")
````

#### <span style="color:#ff66cc; font-weight:bold;">4) DEADLOCKS</span>  

Los `threads` pueden llegar a bloquearse entre si, esperando a varios threads a la vez. Un ejemplo muy sencillo seria un hilo_1 esperando a un hilo_2 y el hilo_2 esperando al hilo_1, osea, ninguno avanzaria.

````python
import threading
import time

lock_A = threading.Lock()
lock_B = threading.Lock()


def hilo1():
    print("Hilo1: Intentando adquirir lock_A...")
    with lock_A:
        print("Hilo1: lock_A adquirido")
        time.sleep(1)  # Simula trabajo
        
        print("Hilo1: Intentando adquirir lock_B...")
        with lock_B:  # ¡Se quedará esperando aquí para siempre!
            print("Hilo1: lock_B adquirido")


def hilo2():
    print("Hilo2: Intentando adquirir lock_B...")
    with lock_B:
        print("Hilo2: lock_B adquirido")
        time.sleep(1)  # Simula trabajo
        
        print("Hilo2: Intentando adquirir lock_A...")
        with lock_A:  # ¡Se quedará esperando aquí para siempre!
            print("Hilo2: lock_A adquirido")


t1 = threading.Thread(target=hilo1)
t2 = threading.Thread(target=hilo2)

t1.start()
t2.start()

t1.join()
t2.join()

print("Programa terminado")  # ¡Nunca llegarás aquí!
````

En este caso, cada thread intenta llegar a una sección con `lock` siendo A y B cada lock, pero los 2 siempre estarán bloqueados hasta que uno se desocupe, cosa que no pasará. 

### <span style="color:#ff5555; font-weight:bold;">3) EJEMPLOS Y APLICACIONES</span>  

#### <span style="color:#ff66cc; font-weight:bold;">1) LOCK COMO ATRIBUTO DE UNA SUBCLASE DE `THREAD`</span>  

Para definir un lock en una clase, debemos definirlo como atributo de la clase y no como objeto, es universal:

````python
import threading, time, os
from random import random

class EscritorArchivo(threading.Thread):
    # Atributo de clase: lo comparten TODAS las instancias
    lock = threading.Lock()

    def __init__(self, id_hilo: int, ruta: str):
        super().__init__(name=f"Hilo-{id_hilo}")
        self.ruta = ruta
        # Cada hilo puede tener estado propio (atributos de instancia)
        self.lineas = 5

    def run(self):
        for i in range(1, self.lineas + 1):
            # Simula trabajo previo a la sección crítica
            time.sleep(0.05 + random() * 0.1)

            # Sección crítica: escribir en el MISMO archivo
            with EscritorArchivo.lock:
                with open(self.ruta, "a", encoding="utf-8") as f:
                    f.write(f"[{self.name}] línea {i}\n")

            # (Fuera del lock, para no retenerlo más de lo necesario)

# Preparar archivo de salida limpio
os.makedirs("files", exist_ok=True)
salida = os.path.join("files", "salida.txt")
open(salida, "w", encoding="utf-8").close()

# Crear y lanzar varios hilos escritores
hilos = [EscritorArchivo(i, salida) for i in range(1, 8)]
for h in hilos: h.start()
for h in hilos: h.join()

print("Listo. Revisa files/salida.txt")
````

#### <span style="color:#ff66cc; font-weight:bold;">2) PATRÓN PRODUCTOR-CONSUMIDOR</span>  

Como una granja de hierro de mc. El buffer recolecta productos generados por un productor y un consumidor los consume. Esto lo haremos con `Deque`, una pila que se puede modificar las 2 partes de ella (agregar, quitar).



````python
import threading
import time
import random
from collections import deque


cofre = deque()
lock = threading.Lock()


def golem_spawner():
    """Produce lingotes de hierro (los golems mueren y dropean hierro)"""
    for i in range(10):
        time.sleep(random.uniform(0.5, 1.5))  # Los golems spawnean aleatoriamente
        lingotes = random.randint(3, 5)  # Cada golem dropea 3-5 lingotes
        
        with lock:
            cofre.append(lingotes)
            print(f"⚔️  Golem #{i+1} murió y dropeó {lingotes} lingotes (En cofre: {len(cofre)})")


def jugador(nombre: str):
    """Consume lingotes de hierro del cofre"""
    inventario = 0
    intentos_vacios = 0
    
    while intentos_vacios < 30:  # Intenta 30 veces antes de rendirse
        with lock:
            if cofre:
                lingotes = cofre.popleft()
                inventario += lingotes
                print(f"👤 {nombre} recogió {lingotes} lingotes (Total: {inventario}, Quedan: {len(cofre)})")
                intentos_vacios = 0
            else:
                intentos_vacios += 1
        
        time.sleep(random.uniform(0.1, 0.3))
    
    print(f"👤 {nombre} terminó con {inventario} lingotes totales")


# Crear la granja (productor)
granja = threading.Thread(target=golem_spawner)

# Crear jugadores (consumidores)
steve = threading.Thread(target=jugador, args=("Steve",))
alex = threading.Thread(target=jugador, args=("Alex",))

print("🏗️  Iniciando granja de hierro...\n")

granja.start()
steve.start()
alex.start()

granja.join()
steve.join()
alex.join()

print("\n✅ Granja completada!")
````

#### <span style="color:#ff66cc; font-weight:bold;">3) QUEUE</span>  

El módulo `queue` nos evita cachos como el anterior donde debemos usar `lock`. Posee métodos `put()` y `get()` que bloquean automaticamente en caso de necesitar lockear (que recalco, lo hacen).

<span style="color:#00cc66; font-weight:bold;">1. put():</span> Agrega un elemento al final de la cola (push)

<span style="color:#00cc66; font-weight:bold;">2. get():</span> Remueve y retorna un item de la cola

<span style="color:#00cc66; font-weight:bold;">3. task_done():</span> Avisa que un item ha sido ya procesado.

<span style="color:#00cc66; font-weight:bold;">4. join():</span> El thread que llame a esto esta en pausa hasta que todos los items sean procesados.


````python
import threading
import time
import random
from queue import Queue


def golem_spawner(cofre: Queue):
    """Produce lingotes de hierro (los golems mueren y dropean hierro)"""
    for i in range(10):
        time.sleep(random.uniform(0.5, 1.5))  # Los golems spawnean aleatoriamente
        lingotes = random.randint(3, 5)  # Cada golem dropea 3-5 lingotes
        cofre.put(lingotes)  # Pone en el cofre
        print(f"⚔️  Golem #{i+1} murió y dropeó {lingotes} lingotes")


def jugador(nombre: str, cofre: Queue):
    """Consume lingotes de hierro del cofre"""
    inventario = 0
    
    while True:
        try:
            lingotes = cofre.get(timeout=3)  # Saca del cofre (espera máx 3 seg)
            inventario += lingotes
            print(f"👤 {nombre} recogió {lingotes} lingotes (Total: {inventario})")
            time.sleep(random.uniform(0.3, 0.8))  # Tiempo procesando
            cofre.task_done()  # ¡Avisa que terminó de procesar este item!
        except:
            print(f"👤 {nombre} terminó con {inventario} lingotes totales")
            break


# El cofre (buffer limitado de 20 slots)
cofre = Queue(maxsize=20)

# Crear la granja (productor)
granja = threading.Thread(target=golem_spawner, args=(cofre,))

# Crear jugadores (consumidores)
steve = threading.Thread(target=jugador, args=("Steve", cofre))
alex = threading.Thread(target=jugador, args=("Alex", cofre))

print("🏗️  Iniciando granja de hierro...\n")

granja.start()
steve.start()
alex.start()

granja.join()  # Espera a que termine de producir
print("\n⏳ Esperando que los jugadores terminen de recoger todo...")
cofre.join()   # Espera a que todos los items tengan task_done()

print("\n✅ ¡Granja completada! Todo el hierro fue recogido")
````
