## EXCEPCIONES EN PYTHON - LEVANTAMIENTO Y MANEJO DE EXCEPCIONES

### 1. EXCEPCIONES EN PYTHON

Condición inesperada que ocurre durante la ejecución de un programa e interrumpe su flujo normal. Esto se llama `excepción`.

Ocurren con problemas como :
- División por cero -> `ZeroDivisionError`
- Leer más allá de los límites de una lista -> IndexError
- Y mas...

Tipos de excepciones más frecuentes

SyntaxError → Error de sintaxis, el código no puede ejecutarse.

if True print("hola")  # SyntaxError


IndentationError → Error en la indentación.

def f():
print("Hola")  # IndentationError


NameError → Variable no definida.

print(x)  # NameError


ZeroDivisionError → División por cero.

10 / 0


IndexError → Índice fuera de rango.

lista = [1, 2]
lista[5]


KeyError → Clave inexistente en diccionario.

d = {"a": 1}
d["b"]  # KeyError


AttributeError → Acceso a atributo inexistente.

x = 5
x.append(1)  # AttributeError


TypeError → Operación con tipos incompatibles.

"hola" + 3  # TypeError


ValueError → Valor inválido para una operación.

int("3x")  # ValueError

### 2. LEVANTAMIENTO Y MANEJO DE EXCEPCIONES

#### `raise`

Python permite generar manualmente una excepción cuando se cumplan ciertas cosas:

```python
def convert_coordinate(coordinate_string):
    if not isinstance(coordinate_string, str):
        raise TypeError("Se esperaba un string")
    if "," not in coordinate_string:
        raise ValueError("Formato inválido, se requiere 'x,y'")
    x, y = coordinate_string.split(",")
    return int(x), int(y)
```

Aqui lanza error si el input no es correcto con el raise, `raise` detiene la ejecución y ejecutará lo que esté escrito.

### `try` y `except`

Sirven para probar código y manejar excepciones correctamente sin que el programa se caiga.

```python
try:
    n = int("34") # modifica esta variable
    print(n)
except ValueError:
    print("No se pudo convertir a número")
```

Puedes generar varios except para prevenir varios errores

```python
try:
    valor = int("abc")
except ValueError:
    print("Error de valor")
except TypeError:
    print("Error de tipo")
```

#### `else` y `finally`

`else`: Se ejecuta solo si no ocurre ningún `except`.

`finally`: Se ejecuta siempre al final.

```python
try:
    print("Intentando división...")
    resultado = 10 / 2
except ZeroDivisionError:
    print("No puedes dividir por cero")
else:
    print("División exitosa:", resultado)
finally:
    print("Este bloque siempre se ejecuta")
```
