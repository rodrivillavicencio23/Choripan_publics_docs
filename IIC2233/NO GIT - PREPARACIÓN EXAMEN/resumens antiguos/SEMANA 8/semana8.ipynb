#### Programación funcional

Las funciones son de primera clase: puedes asignar variables, pasar parámetros o devolver valores. Busca que no existan cambios colaterales y no se ejecuten errores. Entonces nos vamos a que: función: mismos parámetros -> mismo resultado, el retorno solo depende de los argumentos dados a la función, osea que es independiente del programa entero.

#### Función Pura

Siempre entregan el mismo resultado si tiene la misma entrada, además no dependen del resto del programa (esa es la idea del funcional):

```python
def suma(a, b):
    return a + b
```

Una función impura:

```python
x = 10
def suma_impura(y):
    return x + y  # depende de variable externa
# La funcion depende del programa en "x=10"
# Y se obtiene con un input
```

Acá se ve que depende del valor de un x que se encuentra afuera.

#### Consecuencias de la programación funcional

##### Ventajas
1. Código más fácil de probar y razonar
2. Menos errores en programas complejos
3. Ejecución en paralelo
4. No habria sobreescrituta de variables

```python
def aumentar(x: int) -> int:
    return x + 1

print(aumentar(5))  # 6
print(aumentar(5))  # 6 (siempre da lo mismo)
for i in range(10):
    print(aumentar(5))
```

1. Depende solo de los parámetros entregados
2. Por más que ejecute mil veces con un mismo valor "ej: 5", siempre devolverá lo mismo: 6
3. No posee efectos secundarios, no sobreescribe nada, no afecta en ada, ejecuta y retorna

#### Esto al final sirve para:


1. Mayor legibilidad y mantenibilidad
2. Facilidad de testeo
3. Reusabilidad
4. Optimización

### 2. Funciones generadoras

Son funciones especiales que usan `yield`, funciones que recorren datos sin tener que almacenarlos en la memoria. Cuando `iteran` un valor, desaparece hasta que luego de usar completamente la función, el `generador` también desaparece.

```python
def contador():
    n = 1
    while n <= 3:
        yield n
        n += 1

for x in contador():
    print(x)
# 1
# 2
# 3
```

`yield` funciona como un return, solo que no finaliza la función, detiene el programa, manda el `yield` y sigue ejecutandose.

Esto sirve cuando tienes secuencias grandes o infinitas, para no sobrecargar la memoria.

```python
def pares():
    n = 0
    while True:
        yield n
        n += 2

gen = pares()
print(next(gen))  # 0
print(next(gen))  # 2
print(next(gen))  # 4
```

Tambien puedes mandar info a un yield con `.send()`

```python
def eco():
    while True:
        msg = yield
        print("Recibí:", msg)

g = eco()
next(g)           # inicializa el generador
g.send("hola")    # Recibí: hola
g.send("mundo")   # Recibí: mundo
```

#### Iterar con `yield`

Aunque se pueda usar `__iter__` y `__next__`, se puede definir una función usando `yield` que actue como iterador:

```python
def recorrer_lista(lista):
    for item in lista:
        yield item

for elem in recorrer_lista([10, 20, 30]):
    print(elem)
```

### 3. FUNCIONES QUE RETORNAN GENERADORES

Existen 2 tipos: Las dependiendes e independientes de una `función`:

#### Dependientes

1. map:

```python
nums = [1, 2, 3, 4]
dobles = map(lambda x: x*2, nums)
print(list(dobles))  # [2, 4, 6, 8]
```

2. filter:

```python
nums = [1, 2, 3, 4, 5]
pares = filter(lambda x: x % 2 == 0, nums)
print(list(pares))  # [2, 4]
```

3. reduce

```python
from functools import reduce

nums = [1, 2, 3, 4]
suma = reduce(lambda a, b: a + b, nums)
print(suma)  # 10
```

4. funciones lambda

```python
cuadrados = map(lambda x: x**2, [1, 2, 3])
print(list(cuadrados))  # [1, 4, 9]
```

```python
nums = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x**2, nums))
print(cuadrados)  # [1, 4, 9, 16, 25]
```

```python
from functools import reduce
nums = [1, 2, 3, 4, 5]
producto = reduce(lambda x, y: x * y, nums)
print(producto)  # 120
```

#### Independientes

1. enumarate:

```python
letras = ["a", "b", "c"]
for i, letra in enumerate(letras):
    print(i, letra)
# 0 a
# 1 b
# 2 c
```

2. zip:

```python
nombres = ["Ana", "Juan", "Sofía"]
edades = [23, 30, 19]
for nombre, edad in zip(nombres, edades):
    print(nombre, edad)
# Ana 23
# Juan 30
# Sofía 19
```

#### FUNCIONES `built-in`

#### LIBRERIAS `built-in`

itertools: Entrega herramientas a los iteradores y generadores

Ej:

count() : cuenta numeros segun un patron

```python
from itertools import count

contador = count(10, 0.5)

for _ in range(6):
    print(next(contador))
```

cycle() : repite elementos

islice() : recorda iteradores

```python
from itertools import islice, count
for x in islice(count(10), 5):
    print(x)   # 10, 11, 12, 13, 14
```

collections: Proporciona estructuras de datos adicionales como namedtuples deque o defaultdict

Además posee counter: Diccionario especial donde las claves son elementos y los valores son sus frecuencias

```python
from collections import Counter

c = Counter("banana")
print(c)
```

.most_common: Puede darte el valor que mas es frecuente
