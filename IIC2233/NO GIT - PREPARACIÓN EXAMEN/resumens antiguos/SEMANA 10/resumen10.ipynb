### 1. INTERFACES GRAFICAS

Una interfaz gráfica `(GUI)` permite un entorno de interacción usuario - computadora de una forma más amigable con el uso de `ventanas`, `botones`, `menús` y más. Los beneficios de usar una GUI son:
- Interacción más intuitiva
- Mostrar resultados visibles
- Forma normal de un programa

(`historia`) Doug Engelbart junto a su grupo de Stanford creando asi la la `GUI` (Graphical User Interface) cosa que revolucionó la forma de interacción.

#### Arquitectura basada en eventos (EDA)

Es una estructura o patrón de cómo programar, en el caso de los programas GUI usualmente no revisan constantemente si el usuario hizo algo, llamado `polling`, que consume hartos recursos. 

El `EDA` funciona usando cambios o una actualización en el programa como apretar un botón, la reacción a la ejecución de ello se llama `Handler` (Manejador). Para cada evento `e` tendrá una handler `e_handler`.

Uno de los `frameworks` (estructura / marco de trabajo), que trabajan GUI, es `PyQt`.

#### PyQt

Es un `framework` multi-plataforma basado en C++ Qt. 

PyQt posee varias herramientas por lo que esta dividida por partes:
- **QtWidgets**: contiene las clases que brindan los elementos clásicos de interfaces gráficas para aplicaciones en _desktop_ PCs. (Elementos visuales como botones, cuadros...)
- **QtCore**: incluye las clases para funcionalidades no-GUI, como: ciclo de eventos, manejo de archivos, tiempo, _threads_, etc.
- **QtGui**: contiene las classes con componentes para integración de ventanas, manejo de eventos, etc.
- **QtNetwork**: provee las clases para crear aplicaciones gráficas en entornos de red basadas en TCP/IP, UDP.
- **QtOpenGL**: incluye las clases para el uso de OpenGL durante renderizado 3D.
- **QtSvg**: provee de clases para mostrar archivos de gráficos vectoriales (SVG).
- **QtSql**: incluye funcionalidades para el trabajo con bases de datos SQL.
- **QtBluetooth**: contiene clases que permiten la búsqueda e interacción con dispositivos a través de Bluetooth.

#### Crear una ventana
Antes que nada, para toda aplicación gráfica en PyQt se necesita `QApplication` y `QWidget`.
- `QApplication`: Instancia donde se trabaja el ciclo de eventos del sistema. Se necesita para instanciar el resto de widgets ya que sin ella no se mostrarán.
- `QWidget`: clase base de todos los componentes visuales.

Ambas clases pertenecen al módulo de QtWidgets de PyQt5 `PyQt5.QtWidgets`.

En PyQt los elementos básicos que permiten recibir un `evento` y desplegar algo gráficamente se llaman `widgets`.

#### Un ejemplo de una ventana:

```python
import sys
from PyQt5.QtWidgets import QWidget, QApplication


class MiVentana(QWidget):
    def __init__(self) -> None:
        super().__init__()

        # Definimos la geometría de la ventana.
        # Parámetros: (x_superior_izq, y_superior_izq, ancho, alto)
        self.setGeometry(200, 100, 300, 300)

        # Podemos dar nombre a la ventana (Opcional).
        self.setWindowTitle('Mi Primera Ventana')


if __name__ == '__main__':
    def hook(type, value, traceback) -> None:
        print(type)
        print(traceback)
    sys.__excepthook__ = hook

    app = QApplication([])  # Creamos las base de la app: QApplication.
    ventana = MiVentana()   # Construimos un QWidget que será nuestra ventana.
    ventana.show()          # Mostramos la ventana.
    sys.exit(app.exec())    # La aplicación se inicia con app.exec().
                            # Esto habilita el loop de eventos y retorna un
                            # código de salida que luego lo toma "sys.exit()".


#### Otro código:

``` python
import sys
from PyQt5.QtWidgets import QApplication, QWidget

class Ventana(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Ventana inicial")
        # setGeometry(x_superior_izq, y_superior_izq, ancho, alto)
        self.setGeometry(200, 120, 480, 300)

if __name__ == "__main__":
    app = QApplication(sys.argv)
    v = Ventana()
    v.show()
    sys.exit(app.exec_())


### 2. ELEMENTOS GRÁFICOS BÁSICOS PYQT5

#### 2.1 Sistema coordenado en PyQt
El sistema de coordenadas de una ventana de PyQt funciona tal que X crece de izquierda a derecha e Y crece de arriba a abajo, iniciando en la esquina superior izquierda.

Cabe aclarar que cada `widget` tiene una posición y un tamaño definido en píxeles.

##### Ejemplo:

``` python
import sys
from PyQt5.QtWidgets import QWidget, QApplication

class MiVentana(QWidget):
    def __init__(self):
        super().__init__()
        # (x, y, ancho, alto)
        self.setGeometry(200, 100, 300, 300) # (1)
        self.setWindowTitle("Mi Primera Ventana") # (3)

if __name__ == "__main__":
    app = QApplication([])
    ventana = MiVentana()
    ventana.show() # (2)
    sys.exit(app.exec()) # 4

```

- `setGeometry` (1): define la posición y tamaño de la ventana al ejecutarse (x, y, ancho, alto)
- `show()` (2): Muestra la ventana
- `setWindowTitle` (3):  
- `.exec` (4): Inicia el ciclo de eventos

#### 2.3 Etiquetas y cuadros de texto



### 3. EVENTOS Y SEÑALES

#### 3.1 Modelo de eventos
1. Fuente del evento: Objeto que genera el cambio/acción.
2. Objeto evento: Objeto que posee la info del evento.
3. Objeto destino: El objeto que recibe la info y procesa la acción.

PyQt para hacer esto usa el mecanismo de ***signal*** y ***slot***, esto permite la comunicación entre widgets. Emite una signal al slot correspondiente.

Además se pueden usar varias señales para el mismo slot, si usas sender() (QObject) sabrás quién mandó la señal.

#### 3.2 Señales y como detectarlas

```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton, QVBoxLayout

class Ventana(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("sender() demo")

        self.btn_a = QPushButton("A")
        self.btn_b = QPushButton("B")
        self.btn_a.clicked.connect(self.procesar)
        self.btn_b.clicked.connect(self.procesar)

        lay = QVBoxLayout(self)
        lay.addWidget(self.btn_a)
        lay.addWidget(self.btn_b)

    def procesar(self):
        emisor = self.sender()    # <- quién emitió
        print(f"Señal desde: {emisor.text()}")

app = QApplication(sys.argv)
v = Ventana(); v.show()
sys.exit(app.exec())



```

#### 3.3 Eventos de mouse

Existen métodos para la detección de acciones cuando usas un mouse.

1. mousePressEvent(self, event) Cuando se presiona el botón del mouse
2. mouseReleaseEvent(self, event) Cuando se suelta el botón del mouse 
3. mouseMoveEvent(self, event) Posición del mouse

```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel
from PyQt5.QtCore import Qt

class Ventana(QWidget):
    def __init__(self):
        super().__init__()
        self.setMouseTracking(True)               # mover sin presionar
        self.info = QLabel("Mueve/clic en la ventana", self)
        self.info.move(10, 10)

    def mousePressEvent(self, event):
        self.info.setText(f"Press: {event.pos()}")

    def mouseMoveEvent(self, event):
        self.info.setText(f"Move: {event.pos()}") # (x, y) locales

    def mouseReleaseEvent(self, event):
        self.info.setText(f"Release: {event.pos()}")

app = QApplication(sys.argv)
v = Ventana(); v.resize(320, 140); v.show()
sys.exit(app.exec())


```

#### 3.4 Eventos de teclado
Posee eventos casi iguales que el mouse (excepto de movimiento xd):
1. keyPressEvent(self, event)
2. keyReleaseEvent(self, event)

```python

import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel
from PyQt5.QtCore import Qt

class Ventana(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Teclado")
        self.lab = QLabel("Presiona A o ESC", self)
        self.lab.move(10, 10)
        self.setFocus()

    def keyPressEvent(self, event):
        if event.key() == Qt.Key_A:
            self.lab.setText("A presionada")
        elif event.key() == Qt.Key_Escape:
            self.close()

app = QApplication(sys.argv)
v = Ventana(); v.resize(260, 100); v.show()
sys.exit(app.exec())


``` 

#### 3.5 Eventos personalizados 
PyQt también permite definir señales personalizadar por el programador. Para ello se debe crear un objeto que tendrá la nueva señal, que debe ser subclase de `QtCore.QObject`. Dentro de la subclase se crea la nueva señal como instancia de QtCore.pyqtSignal.

1. Usando el método `emit` de `pyqtSignal` se emite la señal.
2. Usando el método `connect` de `pyqtSignal` se define la función a ejecutar con la señal.

#### 3.6 Actualizar UI
Si label.setText(...) no se ve al instante (casos puntuales), fuerza repintado:

label.repaint()

o QApplication.processEvents() cuando haces muchos cambios seguidos

#### 3.7 Señal vs evento
Señales: cuando el propio widget ya expone una señal semántica (ej., clicked, textChanged). Conectar con .connect(...) es lo más claro.

Eventos (override): cuando necesitas datos del suceso (posición, botón del mouse, tecla exacta) o un control más fino del ciclo (mouseMoveEvent, keyPressEvent).

En este capítulo el apunte muestra ambos enfoques: conectar señales (clicked) y sobrescribir manejadores de evento (mouseMoveEvent, keyPressEvent).



### 4. DISEÑO FRONT-BACK

En las aplicaciones gráficas de PyQt se divide en 2 partes:
1. ***Front-end***: Interzas visible donde el usuario interactúa con el programa.
2. ***Back-end***: La lógica del programa, estructura del programa, calculos y más.

Esta forma de programación o diseño de software se guia directamente con ***Alta cohesión*** y ***Bajo acoplamiento***.
- ***Cohesión***: Cada componente debe realizar solo la tarea para la cual fue diseñada.

- ***Acomplamiento***: Cuando la modificación de un componente implica que es necesario modificar a otro componente para lograrlo.

Al separar la ui de la lógica del software se logra este diseño de programación.

Las ventajas de trabajar asi son:
1. Modularidad: Podemos modificar a gusto cada componente sin afectar al otro (Bajo acoplamiento).
2. Uso de recursos: Separa cada proceso del front y back destinando lo necesario a cada uno.
3. Escalamiento: Permite crecer un software de forma más cómoda, sin tener que interferir tanto en procesos ya creados.
4. Experticia: Trabajo en paralelo
5. Mantención: Es posible realizar un testing a gusto.
6. Evolucion del software: Reemplazar o actualizar partes del código manteniendo la misma interfaz.

### 5. CONEXIONES ENTRE MÚLTIPLES VENTANAS EN PYQT


Al crear varias ventanas a la vez uno trataría de poder conectarlas para que funcionen entre sí. Para ello las ventanas no deben crearse dentro de métodos locales sino como atributos de un objeto o conectarse con señales.

```python
from PyQt5.QtWidgets import QApplication, QWidget, QPushButton, QVBoxLayout
from PyQt5.QtCore import pyqtSignal
import sys

class VentanaSecundaria(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Ventana Secundaria")
        self.resize(200, 100)

class VentanaPrincipal(QWidget):
    # Creamos una señal personalizada
    senal_abrir_secundaria = pyqtSignal()

    def __init__(self):
        super().__init__()
        self.setWindowTitle("Ventana Principal")
        self.resize(200, 100)

        # Botón para abrir la otra ventana
        self.boton = QPushButton("Abrir otra ventana")
        self.boton.clicked.connect(self.abrir_ventana)

        layout = QVBoxLayout()
        layout.addWidget(self.boton)
        self.setLayout(layout)

        # Aquí creamos la segunda ventana como atributo (no local)
        self.ventana_secundaria = VentanaSecundaria()

        # Conectamos la señal con el método show() de la otra ventana
        self.senal_abrir_secundaria.connect(self.ventana_secundaria.show)

    def abrir_ventana(self):
        # Emitimos la señal cuando se hace clic
        self.senal_abrir_secundaria.emit()


if __name__ == "__main__":
    app = QApplication(sys.argv)
    ventana_principal = VentanaPrincipal()
    ventana_principal.show()
    sys.exit(app.exec())


```
