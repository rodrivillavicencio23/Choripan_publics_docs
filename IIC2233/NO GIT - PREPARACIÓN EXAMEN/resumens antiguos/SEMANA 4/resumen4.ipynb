### 1. HERENCIA

Es poder heredar los atributos y métodos de una clase, a la que llamaremos `superclase` y a la que hereda `subclase`.
La `subclase` puede seguir funcionando como una clase normal, solo que usando partes de la `superclase`.

Puede:
- Heredar
- Agregar o modificar el comportamiento heredado
- Reutilizar código

Ejemplo:

```python
class Auto:

    def __init__(self, marca, modelo, año, color, km):
        self.marca = marca
        self.modelo = modelo
        self.año = año
        self.color = color
        self._kilometraje = km
        self._dueño = None

    def conducir(self, kms):
        print(f"Conduciendo {kms} kilómetros")
        self._kilometraje += kms

    def vender(self, nuevo_dueño):
        self._dueño = nuevo_dueño
        print(f"Auto vendido a {nuevo_dueño}")

    def leer_odometro(self):
        return self._kilometraje

class Electrico(Auto):

    def __init__(self, marca, modelo, año, color, km, bateria):
        super().__init__(marca, modelo, año, color, km)  ## PUNTO 2
        self.bateria = bateria

    def cargar(self):
        print("Cargando batería...")

    ## PUNTO 1
    def conducir(self, kms):
        print(f"Conducción silenciosa por {kms} kilómetros")
        self._kilometraje += kms
```

Si te fijas en el código, donde dice "## PUNTO 1" ocurre `overriding`, osea la sobreescritura de código para modificarla a gusto.

También está el método llamado `super()` que accede rapidamente a los métodos de la `superclase`. Puede ser el `__init__` (## PUNTO 2) o también podriamos hacer algo como:

```python
def conducir(self, kms):
    print(f"Conducción silenciosa por {kms} kilómetros")
    self._kilometraje += kms
```

También podemos heredar clases de Python ya existentes como `list` o `dict`:

```python
class ListaConLimite(list):
    def __init__(self, limite):
        super().__init__()
        self.limite = limite

    def append(self, item):
        if len(self) < self.limite:
            super().append(item)
        else:
            print("¡Límite alcanzado!")
```

### 2. POLIMORFISMO

Es poder enviar un mensaje con la misma sintaxis (mensaje (método)) a objetos de distintos tipos y que ellos lo adapten a su propia forma. En `python` se logra con `overriding` y `overloading`.

#### OVERRIDING

```python
class Animal:
    def hablar(self):
        print("Sonido genérico")

class Perro(Animal):
    def hablar(self):
        print("Guau!")

class Gato(Animal):
    def hablar(self):
        print("Miau!")

for animal in [Perro(), Gato()]:
    animal.hablar()
# Guau!
# Miau!
```

Ambas clases adaptan el método hablar original de animal y la sobre escriben a su forma.

#### OVERLOADING

Es definir métodos con el mismo nombre pero en `python` no existe esa posibilidad, en lenguajes como `Java` o `C++` si se puede.
Lo que si se puede es re definir operadores (`+`, `-`, `==`) con métodos como `__ad__` `__sub__` `__eq__`.

```python
class Punto:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, otro):
        return Punto(self.x + otro.x, self.y + otro.y)

    def __repr__(self):
        return f"Punto({self.x}, {self.y})"

p1 = Punto(1, 2)
p2 = Punto(3, 4)
print(p1 + p2)  # Punto(4, 6)
```

#### DUCK TYPING

Es una forma de reconocer cosas en programación, da igual que tipo exacto de objeto sea, si tiene métodos/atributos que nos permiten identificar a gran escala que es, seguramente lo sea.
Si camina como un pato y suena como un pato, entonces es un pato 🦆.

```python
class Pato:
    def hablar(self):
        print("Cuac!")

class Persona:
    def hablar(self):
        print("Hola!")

for obj in [Pato(), Persona()]:
    obj.hablar()
```

### 3. MULTIHERENCIA

Es permitir que una clase herede info de más de una clase. Sirve para combinar varias cosas a la vez:

```python
class Animal:
    pass

class Mamifero(Animal):
    pass

class Mascota:
    pass

class Perro(Mamifero, Mascota):
    pass
```

`Perro` hereda tanto de `Mamifero` como de `Mascota`.

Se recomienda usar esto cuando sea realmente necesario, no porque si, ya que puede traer problemas de escritura de código como una mala ejecución al probar el código.

### 4. CLASES ABSTRACTAS

Son clases que no deben instancearse directamente. Sirven para darle un modelo común o estructura a otras clases.

Para que sea considerada una clase como `abstracta` debe heredar `ABC` del modulo abc.

Para que tenga un método que una clase si o si deba tener, se le agrera `@abstractmethod` encima, si no se implementa, tirará error.


Esto es más que nada para ayudar al programador a seguir cierto orden para no equivocarse e implementar todo lo necesario.


```python
from abc import ABC, abstractmethod

class Mamifero(ABC):
    @abstractmethod
    def hablar(self):
        pass

class Perro(Mamifero):
    def hablar(self):
        return "Guau"

class Gato(Mamifero):
    def hablar(self):
        return "Miau"
```

En resumen, es una estructura para el correcto funcionamiento de `subclases` importando el módulo `abc` usando `ABC` y `abstractmethod`.

### 5. DECORADORES

Es una función que recibe otra función (o clase) como entrada y devuelve una modificación de esta.

Se usa para agregar funcionalidades sin modificar el código original (como el `polimorfismo`).

En `Python` se escriben con @nombre_decorador encima de la función/método.



#### OBJETOS DE PRIMERA CLASE


El gran uso de objetos en python es principalmente por:
1. Asignar una función a una variable.
2. Pasar una función como argumento a otra función (o devolverla como resultado)
3. Almacenar funciones en estructuras de datos
4. Posee atributos

```python
# EJEMPLO

# función regular que espera un nombre como string y retorna un string
def decir_hola(nombre: str) -> str:
    return f"Hola {nombre}"

# función regular que espera un nombre como string y retorna un string
def ser_bacan(nombre: str) -> str:
    return f"Oye {nombre}, ¡juntos somos demasiado bacanes!"

# función que espera una función como argumento y retorna un string
def saludar_dani(funcion_saludadora) -> str:
    return funcion_saludadora("Dani")

print(saludar_dani(decir_hola))
print(saludar_dani(ser_bacan))
```

#### SINTAXIS DECORADORES

Como ya vimos el funcionamiento de un `decorador`, ahora tendremos uno aun mas claro:

```python
def mi_decorador(func):
    def nueva_funcion(*args, **kwargs):
        print("Antes de la función")
        resultado = func(*args, **kwargs)
        print("Después de la función")
        return resultado
    return nueva_funcion

@mi_decorador
def saludar():
    print("Hola!")

saludar()
```

#### Decoradores más usados
Los decoradores usan funciones anidadas (funciones que se definen dentro de otra) lo que permite la modificación de esta sin cambiar la definición.


##### @property

```python
class Email:
    def __init__(self, usuario, dominio):
        self.usuario = usuario
        self.dominio = dominio

    @property
    def direccion(self):
        return f"{self.usuario}@{self.dominio}"

mail = Email("juan", "uc.cl")
print(mail.direccion)  # juan@uc.cl

mail = Email("mish", "uc.cl")
print(mail.direccion)
```

Este sirve para la sobre escritura de atributos (mas que nada)

#### @staticmethod

```python
class Matematica:
    @staticmethod
    def suma(a, b):
        return a + b

print(Matematica.suma(3, 4))  # 7
```

Este no depende de un self ni de la clase cls, no depende de atributos ni de la clase en si, solo esta agrupada en una clase.
