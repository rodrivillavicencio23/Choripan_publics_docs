### 1. TUPLAS

Estructuras que permiten agrupar varios valores en un objeto de forma inmutable, osea que una vez creado, no se puede modificar.

```python
coordenadas = (10, 20, 30, 40)
print(coordenadas[0], coordenadas[1])  # 10 20
```

- Se pueden desempaquetar:

```python
coordenadas = (10, 20, 30)
x, y, z = coordenadas
print(x, y, z)  # 10 20 30
```

- Tambien se puede desempaquetar varios valores a la vez usando un asterisco, la posición de este + como sigue el código modificará como se obtendrá cada var:

```python
a, *b, c = (10, 20, 30, 40, 50)
print(a)  # 10
print(b)  # [20, 30, 40]
print(c)  # 50
```

- Aunque no se pueda modificar un dato especifico de una tupla, si el elemento permite la modificación, esta la aceptará, como una lista.

```python
meses = (2023, "semestre", 2, ['Ago', 'Sep', 'Oct', 'Nov', 'Dic'])

meses[3][0] = 'Ene'
print(meses)
```

- Al igual que listas, se puede hacer slicing con ellas:

```python
valores = (1, 2, 3, 4, 5)
print(valores[1:4])  # (2, 3, 4)
```

Tambien Python ofrece la clase `namedtuple` para crear tuplas con nombre:

```python
from collections import namedtuple

Punto = namedtuple("Punto", ["x", "y"])
punto = Punto(10, 20)
print(punto.x, punto.y)  # 10 20
```

### 2. STACKS Y COLAS

#### Un stack (PILAS) es como una pila de platos: el último que entra es el primero que sale (LIFO)

Puedes hacerle:
- Crearla / stack = []
- push / stack.append()
- pop / stack.pop()
- peek (mirar el último elemento) / stack[-1]
- tamaño / len(stack)

#### Colas (Queues)

A diferencia de los stacks, funciona como una fila, el primero que entra, el primero en salir (FIFO)

- enqueue: agrega un elemento al final
- dequeue: saca el primer elemento

Al igual que el anterior, se puede hacer con listas xd

```python
from collections import deque

cola = deque()
cola.append("a")  # enqueue
cola.append("b")
cola.append("c")

print(cola.popleft())  # a (el primero en entrar)
print(cola)     
```

#### Colas dobles

Esto es del jupyter de esta materia, no es tan importante pero por si acaso:

 Implementación en Python: Colas de doble extremo (*deque*)

Un *deque* (pronunciado "*dikiu*" y proviniente de *double-ended queue*) es una estructura secuencial en la que es posible **agregar y sacar elementos desde ambos extremos en forma eficiente**, con un *costo constante por operación*. Esto quiere decir que, independientemente del largo de un *deque*, si éste tiene $N$ elementos, para agregar y sacar elementos siempre ejecutará *la misma* cantidad de operaciones.
Como se explicó anteriormente, esto es mucho mejor que si utilizamos una *lista*, donde la cantidad de operaciones depende de la cantidad de elementos que contiene. En Python, esta estructura es provista por la clase `deque` del módulo `collections` y las principales operaciones que soporta son:

| Operación      | Código Python                | Descripción                                                      |
|----------------|------------------------------|------------------------------------------------------------------|
| Crear *deque*  | `deque()`                    | Crea un *deque* vacío                                            |
| Crear *deque*  | `deque(lista)`               | Crea un *deque* a partir de los elementos de una lista           |
| *Add first*    | `deque.appendleft(elemento)` | Agrega un elemento al inicio del *deque*                         |
| *Add last*     | `deque.append(elemento)`     | Agrega un elemento al final del *deque*                          |
| *Delete first* | `deque.popleft()`            | Retorna y extrae el primer elemento del *deque*                  |
| *Delete last*  | `deque.pop()`                | Retorna y extrae el último elemento del *deque*                  |
| *First*        | `deque[0]`                   | Retorna sin extraer el primer elemento del *deque*               |
| *Last*         | `deque[-1]`                  | Retorna sin extraer el último elemento del *deque*               |
| *length*       | `len(deque)`                 | Retorna el número de elementos en el *deque*                     |
| *Is empty*     | `len(deque) == 0`            | Retorna true si el *deque* está vacío                            |
| *Clear*        | `deque.clear()`              | Limpia el *deque*                                                |
| *Remove*       | `deque.remove(elemento)`     | Saca el primer elemento del *deque* que sea igual a `elemento`   |
| *Count*        | `deque.count(elemento)`      | Cuenta el número de elementos iguales a `elemento` en el *deque* |

El *deque* soporta acceso de lectura y escritura en el elemento de la posición `i`, con la sentencia `deque[i]`. Sin embargo, esta operación **no es eficiente** como en el caso de las listas. En un *deque*, para llegar a la posición `i` el computador inicia en la posición `0` y se va moviendo hasta encontrar la posición `i` para poder leerlo (es decir, requiere recorrer todos los elementos anteriores a `i` para llegar a `i`). 

### 3. DICCIONARIOS

Estructura mutable que no posee una posición tal cual como una lista Ej: list[0]. Sino que funciona a partir de llave -> valor.
Sirve para relacionar información directamente:

```python
monedas = {
    "Chile": "Peso",
    "Perú": "Soles",
    "España": "Euro",
    "Holanda": "Euro",
    "Brasil": "Real"
}

print(monedas["Brasil"])   # Peso
```

Las llaves deben ser inmutables, mientras que los valores se pueden modificar.

```python
dic = {} # crea un diccionario
dic = {"a": 1, "b": 2}
print(dic)

dic["a"] = 10 # modifica
dic["c"] = None # agrega un elemento con su valor

print(dic)
```

Poseen los siguientes métodos:

Tres métodos útiles que existen en los diccionarios son: 

1. `keys()`: permite obtener todas las llaves del diccionario.
2. `values()`: permite obtener todos los valores del diccionario.
3. `items()`: permite obtener cada **par** 'llave-valor" que tiene el diccionario. Cada par es una tupla de la forma `(llave, valor)`.

Además también existe la clase `built-in` llamada `dict` que funciona como un objeto diccionario.

Diccionarios por comprensión o de forma compacta:

```python
cuadrados = {x: x**2 for x in range(5)}
print(cuadrados)  # {0:0, 1:1, 2:4, 3:9, 4:16}
```

### 4. SETS

Contenedor mutable sin orden y sin elementos duplicados.
Los elementos deben ser hashables.

```python
numeros = {3, 2, 3, 3, 1}
print(numeros)  # {1, 2, 3}  (los duplicados desaparecen)

vacio = set()   # set vacío
```

Puedes:
- Agregar / set.add()
- Quitar / set.remove()
- Descartar / set.discard()
- Eliminar todo / set.clear()

La diferencia entre set.remove() y set.discard() es que discard no tirará error si no existe el elemento.

```python
# Crear un set de ejemplo
numeros = {1, 2, 3}
print("Inicial:", numeros)

# 1. .add() → Agregar un elemento
numeros.add(4)
print("Después de add(4):", numeros)

# 2. .remove() → Eliminar un elemento (si no existe → error)
numeros.remove(2)
print("Después de remove(2):", numeros)

# 3. .discard() → Eliminar un elemento (si no existe → no hace nada)
numeros.discard(10)   # 10 no está, pero no da error
print("Después de discard(10):", numeros)
```

### 5. *ARGS - **KWARGS


- `*args`: Pasar una cantidad variable de argumentos según posición 
- `**kwargs`: Pasar una cantidad de variables al estilo diccionario

```python
# *ARGS: solo coloca un asterisco al inicio de la variable
def sumar(*args):
    return sum(args)

print(sumar(1, 2))        # le podrias colocar hartas variables y las junta como lista
print(sumar(1, 2, 3, 4))  
```

```python
# ** KWARGS: usa **var por ejemplo
def mostrar_info(**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

mostrar_info(nombre="Juan", edad=25, ciudad="Santiago")
```

Tambien se pueden desempaquetar

```python
def sumar(x, y, z):
    return x + y + z

nums = [1, 2, 3]
print(sumar(*nums))  # 6

datos = {"x": 4, "y": 5, "z": 6}
print(sumar(**datos))  # 15
```
