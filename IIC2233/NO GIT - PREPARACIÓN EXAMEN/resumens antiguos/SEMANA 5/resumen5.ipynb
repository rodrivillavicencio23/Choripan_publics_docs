## ITERABLES Y GENERADORES - LISTAS LIGADAS

### 1. ITERABLES Y GENERADORES

Los iterables son cualquier elemento que puede recorrerse con un for, como listas, tuplas, strings...

```python
for i in [1, 2, 3]:
    print(i)
```

Mientras que los `iteradores` son los objetos que saben recorrer un iterable.
Un objeto en Python que recuerda en qué posición va de un recorrido y puede darte el siguiente valor cada vez que se le pide.

Tiene que implementar el método especial __iter__() → devuelve a sí mismo.

Tiene que implementar __next__() → devuelve el siguiente elemento o lanza StopIteration cuando no quedan más.

```python
lista = [1, 2, 3]
it = iter(lista)

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
# next(it) → StopIteration

print("---")

lista = [1, 2, 3]
iterador = iter(lista)
next(iterador)
print(next(iterador))
```

#### Generadores

Además poseen su propio `return` que no finaliza al iterador, se usa `yield`:

```python
def contador(n):
    i = 0
    while i < n:
        yield i
        i += 1

for x in contador(3):
    print(x)  # 0, 1, 2
```

`yield` es una función especial, retorna un valor (sin finalizar) y luego borra el valor (se puede guardar en una var) después de ello continua iterando. 

Sirve harto para optimizar almacenamiento y optimización. O incluso para generadores infinitos.

```python
def infinitos():
    i = 0
    while True:
        yield i
        i += 1

gen = infinitos()
print(next(gen))  # 0
print(next(gen))  # 1
print(next(gen))  # 2
```

### 2. LISTAS LIGADAS

#### Nodos

Unidades de información (que ya conozco jaja lol). En `python` se utilizan clases para representar nodos y para representar la conexión se utiliza un atributo como `self.siguiente`.

```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None
```

#### Listas ligadas

Es una especie de lista donde tiene una cabeza y una cola de nodos, como una especie de `grafo`. Basicamente una colección de nodos. 

Para referenciarse no se usan indices, sino referencias usando:
1. valor
2. siguiente
3. next (siguiente nodo)

```python
class ListaLigada:
    def __init__(self):
        self.cabeza = None
```

#### Métodos pinricpales

##### 1. def agregar(valor)

Agrega un nodo al final de la secuencia como si fuera una lista. Se hace de la siguiente forma:
1. Crear el `nodo`
2. Modificar la `cola` del último `nodo` para establecer que hay un siguiente
3. Que el siguiente sea la `cola`

```python
def agregar(self, valor):
    nuevo = Nodo(valor)
    if not self.cabeza:
        self.cabeza = nuevo
    else:
        actual = self.cabeza
        while actual.siguiente:
            actual = actual.siguiente
        actual.siguiente = nuevo
```

#### 2. def obtener(posicion)

Para obtener la posición de un nodo especifico, como no funciona como una lista (no hay indice), se deben recorrer (`iterar`) los nodos desde la `cabeza` hasta el deseado.

```python
def obtener(self, posicion):
    actual = self.cabeza
    indice = 0
    while actual:
        if indice == posicion:
            return actual.valor
        actual = actual.siguiente
        indice += 1
    raise IndexError("Índice fuera de rango")
```

#### 3. def insertar(valor, posicion)

Permite añadir un nuevo `nodo` con su respectivo valor y con una posición. Para ello se puede hacer de tal forma:
1. Crear el nodo con el valor.
2. Recorrer los nodos hasta encontrar el nodo donde queramos añadir el nuevo, al antiguo llamaremos nodo aft.
3. Buscar el nodo predecesor al de la posición que queremos, lo llamaremos nodo pre.
4. Hacer que el nodo nuevo apunte nodo aft.
5. Hacer que el nodo pre apunte al nodo nuevo.


```python
def insertar(self, valor, posicion):
    nuevo = Nodo(valor)
    if posicion == 0:
        nuevo.siguiente = self.cabeza
        self.cabeza = nuevo
        return
    actual = self.cabeza
    indice = 0
    while actual:
        if indice + 1 == posicion:
            nuevo.siguiente = actual.siguiente
            actual.siguiente = nuevo
            return
        actual = actual.siguiente
        indice += 1
    raise IndexError("Índice fuera de rango")
```

Acá un ejemplo con todo el código:

```python
# Clase Nodo
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None


# Clase ListaLigada
class ListaLigada:
    def __init__(self):
        self.cabeza = None

    # 1. Agregar al final
    def agregar(self, valor):
        nuevo = Nodo(valor)
        if not self.cabeza:          # lista vacía
            self.cabeza = nuevo # Asigna al nuevo nodo como cabeza
        else:
            actual = self.cabeza
            while actual.siguiente:  # recorrer hasta que sea None
                actual = actual.siguiente
            actual.siguiente = nuevo # Encuentra al None y lo asigna

    # 2. Obtener por posición
    def obtener(self, posicion):
        actual = self.cabeza # Inicia desde la cabeza
        indice = 0
        while actual:
            if indice == posicion: # Si coincide indice con pos
                return actual.valor # Se encontró el nodo
            actual = actual.siguiente # Sino reasignar la posicion que se recorre
            indice += 1 
        raise IndexError("Índice fuera de rango") # Sino ta malo 

    # 3. Insertar en posición
    def insertar(self, valor, posicion):
        nuevo = Nodo(valor)
        if posicion == 0: # Nueva cabeza
            nuevo.siguiente = self.cabeza
            self.cabeza = nuevo
            return
        actual = self.cabeza
        indice = 0
        while actual:
            if indice + 1 == posicion:
                nuevo.siguiente = actual.siguiente
                actual.siguiente = nuevo
                return
            actual = actual.siguiente
            indice += 1
        raise IndexError("Índice fuera de rango")

    # Para imprimir la lista (extra, útil)
    def __str__(self):
        valores = []
        actual = self.cabeza
        while actual:
            valores.append(str(actual.valor))
            actual = actual.siguiente
        return " → ".join(valores) + " → None"


# -------------------------
# Ejemplo de uso
lista = ListaLigada()
lista.agregar(1)
lista.agregar(2)
lista.agregar(3)
print(lista)          # 1 → 2 → 3 → None

print(lista.obtener(1))  # 2

lista.insertar(99, 1)
print(lista)          # 1 → 99 → 2 → 3 → None
```

Aunque se puedan hacer con un for o while, también se pueden realizar con el nuevo contenido de: `next` y `iter` usandolos como `__next__` y `__iter__`:

```python
from typing import Any
from __future__ import annotations


class IteradorListaLigada:
    def __init__(self, cabeza: Nodo) -> None:
        self.nodo_actual = cabeza

    def __iter__(self) -> IteradorListaLigada:
        return self
    
    def __next__(self) -> Any:
        if self.nodo_actual is None:
            raise StopIteration("Llegamos al final")
        else:
            nodo = self.nodo_actual
            self.nodo_actual = self.nodo_actual.siguiente
            return nodo
```
