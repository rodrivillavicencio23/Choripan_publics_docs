### 1. GUIA DE ESTILO

Reglas de programación que ayuda al trabajo entre programadores, simplemente para que sea más ordenado, limpio y entendible. Ejemplo PEP8 (guia que se usará en programación avanzada IIC-2233 ) 

#### REGLAS DE PEP8
1. Usar "-" en variables si son palabras o usar CamelCase ej: (MuchoGusto)
2. Usar `import` al inicio de un archivo.
3. Usar espacios después de una ", "
4. Usar espacio entre operadores: " 2 + 3 "
5. Usar sangría de 4 tabs
6. Lineas de largo máx de 100 caracteres (incluyendo espacios), sino, separar con \

Ejemplo de uso de PEP8:

```python
# Siempre los imports van al inicio (2)
import math


# Ejemplo de función con nombre en snake_case (1)
def calcular_area_circulo(radio):
    """Devuelve el área de un círculo dado su radio."""
    area = math.pi * (radio ** 2) # (4)
    return area


# Variables con nombres descriptivos en snake_case (1)
radio_circulo = 5
area_resultado = calcular_area_circulo(radio_circulo)

# Uso de espacios en operadores y después de comas (3)
print("El área del círculo con radio", radio_circulo, "es:", area_resultado)

# Ejemplo de línea larga (se corta con \ si supera 100 caracteres) (6)
mensaje_largo = "Este es un mensaje demasiado largo que debería dividirse " \
                "en dos líneas para cumplir con la regla de PEP8"
print(mensaje_largo)
```

### 3. PATHS

Módulo os: Permite usar paths/rutas de archivos 

Los paths pueden ser
- Relativos: Dependen del lugar de la ubicación del archivo ejecutado
- Absolutos: Depende de la ubicación especifica del archivo en el sistema

Los absolutos pueden tener errores de acuerdo al sistema operativo, por eso son mejoers los absolutos.

Código del módulo `os` útil:
- x = os.path.dirname(__file__)
- x = os.path.join(base, subcarpeta, archivo)

```python
import os

# Carpeta + archivo
ruta = os.path.join("carpeta_path", "archivo.txt")

with open(ruta, "r", encoding="utf-8") as f:
    a = f.read()

print(ruta)
print(a)
```
