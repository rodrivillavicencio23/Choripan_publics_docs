### 1. THREADING

Para solucionar problemas del como en un juego donde ocurren muchos eventos a la vez se complica un poco.En `python` podriamos pensar sobre el uso de varios ***inputs***, cosa que es incómodo, por ello nace el `thread`, este nos permite definir múltiples flujos de instrucciones dentro del mismo código de neustro programa

#### 1.1 Procesos y núcleos

Un computador (sistema operativo) puede ejecutar varios programas o aplicaciones a la vez gracias a **procesos**. Un proceso es un **programa en ejecución** que ocupa **RAM** del computador y ejecuta un flujo de instrucciones. 

La unidad que permite la ejecución de programas es la **CPU** que posee **cores** que lee las instrucciones del programa, la ejecuta y sigue. Esto mismo ocurre con Python cuando ejecutamos nuestro código.

Los dispositivos electrónicos modernos suelen tener incluso más de un **núcleo** (**core**) pudiendo ser dual, quad u octa. 

Incluso para la ejecución de una tarea, los cores se podrían repartir el trabajo entre ellos para realizarlo. Esto por ejemplo, ya es parte de lo que se puede hacer con `threads`.



#### 1.2 Paralelismo Vs Concurrencia

- Paralelismo: Ejecución de varias tareas en diferentes núcleos o procesadores.
- Concurrencia: Ejecuta varias tareas de forma intercalada en el mismo núcleo o procesador.

En `Python` se puede realizar solo **Concurrencia** con el módulo `threading`. Mientras que **Paralelismo** no se puede hacer de forma nativa, se puede realizar con otra librería llamada `multiprocessing`. 

(El ***paralelismo*** para que se pueda realizar en Python puede usar ***multiprocessing**, en este caso como Python es un lenguaje interpretado, permite que se pueda usar otro lenguaje donde si se permite el paralelismo en si y usarlo)

```

#### 1.3 **Threading**

Un **thread** o hilo, es una unidad de ejecucicón de código dentro de un proceso. Todo programa tiene al menos 1 **thread** (hilo), llamado **thread principal**, este se encarga de recorrer el código principal del programa y ejecutar las instrucciones. Igualmente pueden existir más de 1 hilo para ejecutar diferentes partes de un código al mismo tiempo.

Cada **hilo** mantiene sus variables locales y posición respecto en el código (tipo estamos en la linea 31), por lo que cada **thread** funciona de forma independiente. También un hilo puede crear otros.

Cada hilo posee un tiempo asigando de la CPU para ejecutarse. Esto ocurre gracias a la ***planificación de hilos*** (thread scheduling o time slicing), funciona turnando hilos de la siguiente forma :
1. Se selecciona un hilo a ejecutar.
2. Se ejecuta por el tiempo asignado.
3. Se detiene y se pone en espera.
4. Se ejecuta otro hilo, repitiendo el ciclo.

Cuando el computador tiene solo 1 **núcleo** los hilos no se ejecutan simultaneamente, sino que alternan de forma veloz para lograr esto. Aunque esto sea más ordenado y permite separar diferentes tareas con mayor orden, este igualmente consume recursos del computador.



#### 1.4 Uso de threads

Estos son útiles cuando queremos que sse realicen varias tareas sin que se interpongan. Un ejemplo de esto es un hilo que descarga información por internet y otro que procesa lo descargado.

También sirve para programar que deben mantener interfaces activas en segundo plano.

Asi el flujo del programa no se detiene por completo mientras ocurren otros procesos.

`IMPORTANTE:` Los threads son de 1 solo uso, osea se consumen.

#### 1.5 Módulo `threading`

Como ya se mencionó previamente, el **módulo** `threading` permite trabajar con hilos en ***Python***. Proporciona clases y funciones para trabajar con hilos. 
1. Thread(target=función): crea un hilo nuevo que ejecutará la función indicada.
2. start(): inicia la ejecución del hilo.
3. join(): bloquea el hilo principal hasta que el hilo secundario termine.
4. is_alive(): indica si un hilo sigue corriendo.
5. current_thread(): devuelve el hilo actual.

#### 1.8 Códigos de `threads`

##### **join()**

Se utiliza para que el programa principal espere hasta que un **hilo** termine su ejecución antes de continuar. Permite un mejor flujo del programa:

````python
import threading
import time

def tarea_lenta():
    print("Inicio de tarea lenta...")
    time.sleep(3)
    print("Tarea lenta terminada.")
    time.sleep(1)

hilo = threading.Thread(target=tarea_lenta)
hilo.start()

# Espera a que el hilo termine antes de continuar
hilo.join() # PRUEBA QUITANDO ESTA LINEA
print("El programa continúa después del hilo.")

````

##### **is_alive()**

Sirve para verificar si un hilo sigue activo. Permite monitorear tareas o sincronizar partes del programa con otros hilos:

````python
import threading
import time

def tarea():
    time.sleep(2)
    print("Hilo finalizado.")

hilo = threading.Thread(target=tarea)
hilo.start()

while hilo.is_alive():
    print("El hilo sigue ejecutándose...")
    time.sleep(0.5)

print("El hilo terminó completamente.")
````

Mientras que **threads normales** necesitan finalizar para detener el programa, los **daemon** finalizan automaticamente si el programa finaliza. 

Este es un ejemplo de un código **daemon**. `NO EJECUTAR EN UN JUPYTER NOTEBOOK`

### 2. CONCURRENCIA Y CONSECUENCIAS

#### 2.1 SINCRONIZACIÓN

##### 2.2 ***locks***

Cuando varios ***threads*** comparten datos, como un contador, pueden producirse mezclas con variables y otros datos si estos modifican exacatamente los mismos parámetros. Para evitar resultados incorrectos hay que proteger la sección "vulnerable". Para solucionar esto se introducen 2 mecanizmos principales: ***locks*** y ***events***.

Un **lock** (candado) se comparte entre **threads** y permite que solo uno entre a esta sección vulnerable o crítica. Posee un estado binario de bloqueado/desbloqueado. Para activarlo se hace usando `acquire()` antes de la sección y `release()` al salir.

```python

        import threading

        lock_global = threading.Lock()

        def sumador_con_seccion_critica(contador: Contador, lock: threading.Lock) -> None:
            for _ in range(10**6):
                lock.acquire()
                # --- sección crítica ---
                contador.valor += 1
                # --- fin sección crítica ---
                lock.release()

        contador = Contador()
        t1 = threading.Thread(target=sumador_con_seccion_critica, args=(contador, lock_global))
        t2 = threading.Thread(target=sumador_con_seccion_critica, args=(contador, lock_global))

        t1.start(); t2.start()
        t1.join(); t2.join()
        print("Listo, nuestro contador vale", contador.valor)


```

El **lock** debe ser el mismo objeto para todos los hilos, si cada hilo crea su propio lock no protege nada:

```python

        import threading

        def sumador_con_seccion_critica_mala(contador: Contador) -> None:
            lock = threading.Lock()  # ¡Cada hilo crea su lock! (mal)
            for _ in range(10**6):
                lock.acquire()
                contador.valor += 1
                lock.release()

        contador = Contador()
        t1 = threading.Thread(target=sumador_con_seccion_critica_mala, args=(contador,))
        t2 = threading.Thread(target=sumador_con_seccion_critica_mala, args=(contador,))

        t1.start(); t2.start()
        t1.join(); t2.join()
        print("Listo, nuestro contador vale", contador.valor)


```




##### 2.3 Locks y Context Managers 

También podemos hacer que locks puedan usar `with` para asegurar que se liberen incluso si ocurre una excepción. Esto lo hace sin `acquire()` y `release()`.

```python
        import threading, time

        lock_global = threading.Lock()

        def sumador(contador: Contador, lock: threading.Lock) -> None:
            nombre = threading.current_thread().name
            for _ in range(10):
                with lock:
                    valor = contador.valor
                    print(f"{nombre}: lee {valor}")
                    nuevo_valor = valor + 1
                    time.sleep(0.05)  # simula trabajo real
                    print(f"{nombre}: suma 1 => {nuevo_valor}")
                    contador.valor = nuevo_valor
                    print(f"{nombre}: guarda {contador.valor}")

        contador = Contador()
        t1 = threading.Thread(name="T1", target=sumador, args=(contador, lock_global))
        t2 = threading.Thread(name="T2", target=sumador, args=(contador, lock_global))
        t1.start(); t2.start()
        t1.join(); t2.join()
        print("Valor final protegido con lock:", contador.valor)


```

#### 2.4 Señales entre threads: Event

Aunque tenemos `join()` para la interacción en cadena entre `threads` se necesita en algunas veces que ocurra algo en particular más que simplemente que se finalice el `thread`. Para eso esta `Event`. Al igual que `lock` un `Event` posee un carácter binario, pero este lo que hace es sincronizar el momento en que un hilo puede avanzar.

Un ejemplo es un video, este tiene imagen y audio, el video no empezará si no carga su imagen y tampoco su audio, deben sincronizarse para empezar, para ello podria usar `event()`.

Un `event()` posee:
1. `.set()` activa el evento
2. `.wait()` hace que el hilo espere hasta que se active el evento
3. `.clear()` coloca el evento en `false`
4. `.is_set()` permite consultar si esta activado



````python
import threading, time

video_listo = threading.Event()
audio_listo = threading.Event()

def cargar_video():
    print("Cargando video (10 segundos)...")
    time.sleep(10)
    print("Video cargado. Esperando audio...")
    video_listo.set()          # avisa que el video está listo
    audio_listo.wait()         # espera a que el audio también esté listo
    print("Reproduciendo video + audio")

def cargar_audio():
    print("Cargando audio (3 segundos)...")
    time.sleep(3)
    print("Audio cargado. Esperando video...")
    audio_listo.set()          # avisa que el audio está listo
    video_listo.wait()         # espera a que el video esté listo
    print("Reproduciendo audio + video")

t1 = threading.Thread(target=cargar_video)
t2 = threading.Thread(target=cargar_audio)
t1.start(); t2.start()
t1.join(); t2.join()
````

#### 2.5 Deadlocks

Puede que cuando algunos threads interactuan entre si pueden quedarse pillados o atrapados sin que ninguno avance. A esto se le llama ***deadlock*** o ***interbloqueo***. Por ejemplo este:

```python
        import threading, time

        lock_1 = threading.Lock()
        lock_2 = threading.Lock()

        def master():
            print("Master: adquiriendo lock_1")
            with lock_1:
                time.sleep(0.5)
                print("Master: adquiriendo lock_2")
                with lock_2:
                    print("Master: terminó")

        def worker():
            print("Worker: adquiriendo lock_2")
            with lock_2:
                time.sleep(0.5)
                print("Worker: adquiriendo lock_1")
                with lock_1:
                    print("Worker: terminó")

        t_master = threading.Thread(target=master)
        t_worker = threading.Thread(target=worker)
        t_master.start(); t_worker.start()
        t_master.join(); t_worker.join()


```

#### 3.1 ***Lock*** como atributo de una subclase de `thread`

Si tenemos varias instancias de una  **Thread** y comparten un recurso en común, como un archivo, deben compartir un mismo ***lock*** para proteger la sección crítica. Para hacer esto podemos establecer el ***lock*** como atributo de la `Clase` (Osea fuera del `__init__`). De esta forma, todos poseeran el mismo candado.

```python
        import threading, time, os
        from random import random

        class EscritorArchivo(threading.Thread):
            # Atributo de clase: lo comparten TODAS las instancias
            lock = threading.Lock()

            def __init__(self, id_hilo: int, ruta: str):
                super().__init__(name=f"Hilo-{id_hilo}")
                self.ruta = ruta
                # Cada hilo puede tener estado propio (atributos de instancia)
                self.lineas = 5

            def run(self):
                for i in range(1, self.lineas + 1):
                    # Simula trabajo previo a la sección crítica
                    time.sleep(0.05 + random() * 0.1)

                    # Sección crítica: escribir en el MISMO archivo
                    with EscritorArchivo.lock:
                        with open(self.ruta, "a", encoding="utf-8") as f:
                            f.write(f"[{self.name}] línea {i}\n")

                    # (Fuera del lock, para no retenerlo más de lo necesario)

        # Preparar archivo de salida limpio
        os.makedirs("files", exist_ok=True)
        salida = os.path.join("files", "salida.txt")
        open(salida, "w", encoding="utf-8").close()

        # Crear y lanzar varios hilos escritores
        hilos = [EscritorArchivo(i, salida) for i in range(1, 8)]
        for h in hilos: h.start()
        for h in hilos: h.join()

        print("Listo. Revisa files/salida.txt")


#### 3.2 Patrón Productor-consumidor 

Un **productor*** genera ítems en un **buffer** y un ***consumidor*** los toma. Esto permite la comunicación entre threads usando colas sincronizadas o una cola thread safe. `Deque` Nos permite agregar y sacar elementos en ambos extremos pero nada nos asegura que salte un error, por lo que usamos locks.

Ejemplo de un panadero y compradores

``` python
        from collections import deque
        from random import choices, randint
        import threading, time

        piezas_de_pan = deque()
        lock = threading.Lock()
        PANES = ["Marraqueta", "Baguette", "Hallulla"]

        def panadero():
            for partida in range(3):
                time.sleep(2)
                lote = choices(PANES, k=5)
                with lock:
                    piezas_de_pan.extend(lote)
                    print(f"[Panadero] Partida {partida+1}: {lote} (total={len(piezas_de_pan)})")

        def cliente(i):
            time.sleep(randint(0, 5) / 10)
            while True:
                with lock:
                    if piezas_de_pan:
                        pan = piezas_de_pan.popleft()
                        print(f"[Cliente {i}] Compré {pan} (restan={len(piezas_de_pan)})")
                        return
                time.sleep(0.05)

        tp = threading.Thread(target=panadero)
        tc = [threading.Thread(target=cliente, args=(i,)) for i in range(15)]

        tp.start()
        for t in tc: t.start()
        for t in tc: t.join()
        tp.join()


```

Esto no posee erroes pero se vuelve un código más complejo.

Para esto se recomienda usar `Queue` que es trhead-safe.


#### 3.3 Queu

El *módulo* `Queue` promete una cola segura para trabajar con hilos. Posee métodos *put()* y *get()* que bloquean **automaticamente** si es necesario y manejan sus propios ***locks***.
1. `put()`: Agrega un elemento al final de la cola (*push*)
2. `get()`: Remueve y retorna un ítem de la cola. Este método espera que haya algo en la cola para sacarlo.
3. `task_done()`: Requiere ser llamado cada que un ítem estraído de la cola ha sido procesado.
4. `join()`: El thread que llame a este método queda en pausa hasta que todos los ítems de la cola hayan sido procesados.

```python
        from queue import Queue
        from random import choices, randint
        import threading, time

        PANES = ["Marraqueta", "Baguette", "Hallulla"]
        cola = Queue()
        N_CONSUMIDORES = 5
        CENTINELA = None  # marca de fin

        def panadero():
            for partida in range(3):
                time.sleep(2)
                lote = choices(PANES, k=5)
                for pan in lote:
                    cola.put(pan)
                print(f"[Panadero] Partida {partida+1}: {lote}. Total en cola: {cola.qsize()}")
            for _ in range(N_CONSUMIDORES):
                cola.put(CENTINELA)

        def cliente(i):
            while True:
                pan = cola.get()
                if pan is CENTINELA:
                    cola.task_done()
                    print(f"[Cliente {i}] Me voy (no queda pan).")
                    break
                time.sleep(randint(0, 3) / 10)
                print(f"[Cliente {i}] Compré {pan}. Pendientes en cola: {cola.qsize()}")
                cola.task_done()

        threads = [threading.Thread(target=cliente, args=(i,)) for i in range(N_CONSUMIDORES)]
        for t in threads: t.start()

        panadero_thread = threading.Thread(target=panadero)
        panadero_thread.start()

        panadero_thread.join()
        cola.join()
        print("Cerrado ordenado.")


```



#### 3.4 Simulación con event

Utiliza lo visto anteriormente pero con event.

```python
        from threading import Event, Thread
        from time import sleep
        from random import randint

        class Estudiante(Thread):
            def __init__(self, lineas_estudio: int, senal_duda: Event, senal_resuelto: Event):
                super().__init__(name="Estudiante")
                self.max_lineas = lineas_estudio
                self.senal_duda = senal_duda
                self.senal_resuelto = senal_resuelto

            def run(self):
                for linea in range(1, self.max_lineas + 1):
                    print(f"[{self.name}] Programando línea {linea}...")
                    sleep(randint(1, 3) / 2)
                    if randint(1, 5) == 3:
                        print(f"[{self.name}] ¡Profe, tengo una duda!")
                        self.senal_resuelto.clear()
                        self.senal_duda.set()
                        self.senal_resuelto.wait()
                        print(f"[{self.name}] ¡Gracias, sigo!")

        class Profe(Thread):
            def __init__(self, senal_duda: Event, senal_resuelto: Event):
                super().__init__(name="Profe")
                self.senal_duda = senal_duda
                self.senal_resuelto = senal_resuelto

            def run(self):
                while True:
                    self.senal_duda.wait()
                    self.senal_duda.clear()
                    print("[Profe] A ver, muéstrame...")
                    sleep(randint(1, 4) / 2)
                    print("[Profe] Listo, resuelto.")
                    self.senal_resuelto.set()

        senal_tengo_duda = Event()
        senal_resuelta = Event()

        est = Estudiante(10, senal_tengo_duda, senal_resuelta)
        profe = Profe(senal_tengo_duda, senal_resuelta)

        profe.daemon = True
        profe.start()
        est.start()
        est.join()
        print("Clase terminada.")
