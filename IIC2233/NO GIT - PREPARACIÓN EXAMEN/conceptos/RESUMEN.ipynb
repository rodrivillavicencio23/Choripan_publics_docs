## <span style="color:#ff5555; font-weight:bold;">INTERFACES GRÁFICAS</span>

### <span style="color:#f1c40f; font-weight:bold;">GUI:</span> Interfaces gráficas. El usuario interactua comodamente a través de pantallas con un programa (No con la consola).

#### <span style="color:#00cc66; font-weight:bold;">POLLING:</span> Revisar activamentes si el estado de algo ha cambiado en un programa.

#### <span style="color:#00cc66; font-weight:bold;">ARQUITECTURA BASADA EN EVENTOS:</span> El código reacciona a eventos (como un botón).

### <span style="color:#f1c40f; font-weight:bold;">PYQT:</span> Libreria de Python que permite trabajar con un `framework` para hacer `GUI`.

#### <span style="color:#00cc66; font-weight:bold;">WIDGETS:</span> Elementos de una `GUI` que construyen el entorno. 

##### <span style="color:#ff8800; font-weight:bold;">VENTANA:</span> Widget raiz que se encarga de manejar y mostrar widgets.

##### <span style="color:#ff8800; font-weight:bold;">ETIQUETAS Y CUADROS DE TEXTO:</span> Las `QLabel` (Etiquetas) muestran texto y las `QLineEdit` (Cuadros de texto) reciben texto del usuario.

##### <span style="color:#ff8800; font-weight:bold;">IMAGENES:</span> Se pueden agregar usando QPixMap del módulo QtGui. Depende de una `QLabel` para ser agregado.

##### <span style="color:#ff8800; font-weight:bold;">BOTONES:</span>: Para agregarlos se usan `QPushButton`.

##### <span style="color:#ff8800; font-weight:bold;">LAYOUTS:</span> Manejan la distribución de widgets en la ventana más comodamente. Usan las `box layouts` que manejan según horizontal y vertical, un Box para cada uno. `QHBoxLayout` y `QVBoxLayout` de `QtWidgets`.

##### <span style="color:#ff8800; font-weight:bold;">GRID LAYOUT:</span> Al igual que la anterior pero lo divide por casillas usando filas y columnas. 

#### <span style="color:#00cc66; font-weight:bold;">POO para widgets:</span> Todo elemento de PyQT es un objeto, por lo que se les puede tratar como uno. Se les puede hacer override por ejemplo.

### <span style="color:#f1c40f; font-weight:bold;">EVENTOS Y SEÑALES:</span> PyQT usa eventos y señales para funcionar. Esto una vez cuando se ejecute el programa (metodo `exec()` de `QApplication`)

#### <span style="color:#00cc66; font-weight:bold;">PILARES DE UN EVENTO:</span> `1)` Fuente del evento; `2)` Objeto del evento (info del evento); `3)` Objeto del destino 

##### <span style="color:#ff8800; font-weight:bold;">SENDER:</span> Método para saber quién envió una señal.

#### <span style="color:#00cc66; font-weight:bold;">TIPOS DE EVENTOS:</span> Existen diversos objetos que pueden hacer eventos:

##### <span style="color:#ff8800; font-weight:bold;">MOUSE:</span>

- `mousePressEvent(self, event):` Cuando se presiona el botón del mouse
- `mouseReleaseEvent(self, event):` Cuando suelta el botón del mouse
- `mouseMoveEvent(self, event):` Pos del mouse

##### <span style="color:#ff8800; font-weight:bold;">TECLADO:</span>

- `keyPressEvent(self, event)`
- `keyReleaseEvent(self, event)`

##### <span style="color:#ff8800; font-weight:bold;">EVENTOS PERSONALIZADOS:</span> Se pueden crear objetos que tengan la nueva señal como atributo (debe heredar `Qt.Core:QObject`) La nueva señal se crea como instancia de `QTCore.pyqtSignal`.

##### <span style="color:#ff8800; font-weight:bold;">SEÑAL VS EVENTO:</span> 

- `Señales:` Las señales vienen por default.
- `Evento:` Se les debe hacer override.

### <span style="color:#f1c40f; font-weight:bold;">FRONTEND Y BACKEND:</span>

- `FRONTEND:` Intertaz visible que tiene la interacción usuario-computador.
- `BACKEND:` Lógica interna del programa.

### <span style="color:#f1c40f; font-weight:bold;">ALTA COHESIÓN Y BAJO ACOPLAMIENTO:</span>

- `Cohesión:` Los elementos no se afectan entre si.
- `Acoplamiento:` Los módulos no dependen entresi.

Esto logra:

<span style="color:#ff8800; font-weight:bold;">1. Modularidad</span>  
<span style="color:#ff8800; font-weight:bold;">2. Escalamiento</span>  
<span style="color:#ff8800; font-weight:bold;">3. Mantención</span>  
<span style="color:#ff8800; font-weight:bold;">4. Experticia</span>  
<span style="color:#ff8800; font-weight:bold;">5. Actualización</span>  
<span style="color:#ff8800; font-weight:bold;">6. Uso de recursos</span>  

Si se hace un buen `front-back` se logra los 2 puntos anteriores.

### <span style="color:#f1c40f; font-weight:bold;">CONEXIÓN ENTRE VENTANAS:</span> Para que varias ventanas interactuen entre si, se debe crear una señal personalizada.

## <span style="color:#ff5555; font-weight:bold;">THREADING</span>  

### <span style="color:#f1c40f; font-weight:bold;">THREADS:</span> Tambien llamados hilos. Son flujos de programa, todo programa tiene un flujo pero pueden llegar a tener más de uno. Estos pueden ejecutar código de forma independiente al `thread principal`.

### <span style="color:#f1c40f; font-weight:bold;">PROCESOS Y NÚCLEOS:</span> Un computador posee su ***CPU*** donde se ejecutan procesos. Los ***CPU*** pueden tener varios núcleos donde se ejecutan subprocesos, o flujo de instrucciones, de manera independiente.

### <span style="color:#f1c40f; font-weight:bold;">PARALELISMO Y CONCURRENCIA:</span>
<span style="color:#00cc66; font-weight:bold;">1. PARALELISMO:</span> Es la ejecucicón de diversos flujos de instrucciones gracias al uso de distintos núcleos de un procesador.

<span style="color:#00cc66; font-weight:bold;">2. CONCURRENCIA</span> Ejecuta varias tareas intercaladamente, simulando un aparente `paralelismo` por su velocidad, pero este usa un núcleo solamente.

### <span style="color:#f1c40f; font-weight:bold;">MÓDULO `threading`:</span> Se encarga de crear `threads` en Python. Gracias al funcionamiento interno de Python, <span style="color:#ff8800; font-weight:bold;">no se puede realizar paralelismo, solo concurrencia</span> por lo que al usar threads, se van intercalando su turno velozmente.  

#### <span style="color:#00cc66; font-weight:bold;">CÓDIGO BÁSICO:</span> 

<span style="color:#ff8800; font-weight:bold;">1. .Thread(target = func) </span> 

<span style="color:#ff8800; font-weight:bold;">2. .start() </span>

<span style="color:#ff8800; font-weight:bold;">3. .join() </span>

<span style="color:#ff8800; font-weight:bold;">4. .is_alive()</span>

<span style="color:#ff8800; font-weight:bold;">5. .current_thread()</span>   

#### <span style="color:#00cc66; font-weight:bold;">DAEMON THREADS:</span> Los threads comunes finalizan su trabajo cuando terminan su función. Los `daemon` lo hacen solo cuando el programa principal finaliza.

<span style="color:#ff8800; font-weight:bold;">1. daemon1 = threading.Thread(name = "Daemon1", target = func, daemon = True)</span>  

<span style="color:#ff8800; font-weight:bold;">2. daemon2 = threading.Thread(name = "Daemon2", target = func)    dameon2.daemon = True </span> 

 

#### <span style="color:#00cc66; font-weight:bold;">CLASES EN THREADS:</span> En vez de solo tener funciones, tambien se le pueden colcoar clases e incluso hacerle `override`.

#### <span style="color:#00cc66; font-weight:bold;">TIMERS:</span> A los `threads` se les puede colocar timers para que luego de cierto rato, ejecuten su código.

### <span style="color:#f1c40f; font-weight:bold;">ADVERTENCIA:</span> Hay que tener cuidado que los `threads` pueden llegar a superponerse y cambiar variables o no dejar resultados deseados.

Por ello existen los `locks`.

#### <span style="color:#00cc66; font-weight:bold;">LOCKS:</span> Cerraduras que bloquean cierta parte de código para que todo funcione como debería. Esto sirve para acceder a secciones <span style="color:#ff66cc; font-weight:bold;">críticas</span>  en un código (entendiendo zonas <span style="color:#ff66cc; font-weight:bold;">críticas</span> como un espacio que varios `threads` pueden modificar entre si) 

Se pueden activar con `acquire()` y desactivar con `release()`. Tambien se pueden ejecutar usando with `lock`:

## <span style="color:#ff5555; font-weight:bold;">NETWORKING</span>  

### <span style="color:#f1c40f; font-weight:bold;">SERIALIZACIÓN DE OBJETOS:</span> Transformar un objeto en una serie de `bytes`. 

### <span style="color:#f1c40f; font-weight:bold;">JSON:</span> Formato de texto interpretado por muchos lenguajes.

En python existe la libreria `json` que puede serializar y deserializar en `str` o en un archivo. Para <span style="color:#ff8800; font-weight:bold;">archivo</span>  se usa `dump` y `load` y para <span style="color:#ff8800; font-weight:bold;">str</span>  se usa `dumps` y `loads`.

### <span style="color:#f1c40f; font-weight:bold;">COMUNICACIÓN ENTRE COMPUTADORES:</span> Para que 2 máquinas (`hosts`) se comuniquen necesitan:

<span style="color:#00cc66; font-weight:bold;">1. Un medio</span>  

<span style="color:#00cc66; font-weight:bold;">2. Destinatario y como encontrarlo.</span>  

<span style="color:#00cc66; font-weight:bold;">3. Un protocolo de transmisión de mensajes.</span>  

<span style="color:#00cc66; font-weight:bold;">4. Un protocolo de aplicación.</span>  

#### <span style="color:#ff66cc; font-weight:bold;">IP Y PORTS</span> 

Todo `host` tiene una **etiqueta** llamada `dirección ip` (como el rut). Existen 2 tipos, los de 4 bytes (`IPv4`) y los de 6 bytes (`IPv6`)

Además cada `host` tiene muchos `ports`. Puertas que permite conectar diferentes conexiones.

#### <span style="color:#ff66cc; font-weight:bold;">PROTOCOLO DE COMUNICACIÓN:</span> Para los mensajes, se dividen en mini paquetes para que sa más fácil. Existen 2 principales:

#### <span style="color:#ff8800; font-weight:bold;">TCP:</span> Lento pero seguro. Tiene `handshake`.

#### <span style="color:#ff8800; font-weight:bold;">UDP:</span> Rápido pero ocurren muchos errores de envío. No tiene `handshake`

### <span style="color:#f1c40f; font-weight:bold;">ARQUITECTURA CLIENTE-SERVIDOR:</span> Estructura de ***software*** que busca que 2 programas o más se conecten y uno sea un `servidor` que ofrece servicios y el otro que sea `cliente` que consume los servicios.

#### <span style="color:#ff66cc; font-weight:bold;">WEBSERVICES:</span> Aplicaciones que utilizan el formato `cliente-servidor`.

#### <span style="color:#ff66cc; font-weight:bold;">API:</span> Conjunto de normas y funciones que permite la comunicación `cliente-servidor`. Tiene una `URL BASE` y también `ENDPOINTS`, rutas que nos da la `API` para hacer consultas (que pueden ser personalizadas con `parámetros`). 

#### <span style="color:#ff66cc; font-weight:bold;">HTTP:</span> Protocolo de comunicación ***cliente-servidor***. Funciona en base al protocolo ***request-respone***.

Tiene los métodos principales: <span style="color:#00cc66; font-weight:bold;">GET; POST; PATCH; PUT; DELETE.</span>  

`HTTP` responde a las solicitudes principalmente según un valor númerico. <span style="color:#00cc66; font-weight:bold;">{200: TODO OK}; {403: RECHAZADO}; {404: NO ENCONTRADO}; {500: ERROR DEL SERVIDOR}</span>  

#### <span style="color:#ff66cc; font-weight:bold;">CLIENTE:</span> `Host` que funciona como cliente en el rol `cliente-servidor`. Pide solicitudes al servidor con la librería `request`.

##### <span style="color:#ff8800; font-weight:bold;">MÉTODOS:</span>  Los mencionados anteriormente: `GET; POST; PATCH; PUT; DELETE`.

<span style="color:#00cc66; font-weight:bold;">1. GET: Obtiene datos</span>  `| No autorización`

<span style="color:#00cc66; font-weight:bold;">2. POST: Crea recursos en el servidor</span> `| No autorización`

<span style="color:#00cc66; font-weight:bold;">3. PATCH: Modifica parcialmente un recurso </span> `| Necesita autorización`

<span style="color:#00cc66; font-weight:bold;">4. PUT: Reemplaza un recurso</span> `| Necesita autorización`

<span style="color:#00cc66; font-weight:bold;">5. DELETE: Elimina un recurso</span> `| Necesita autorización`
